3.Constructor (__init__) :
   
3.1 Understand the Concept

A constructor is a special method that is automatically called when an object is created.
In Python, the constructor is commonly written using __init__().

It is used to initialize the object's data.

Example / Demonstration :

Suppose every ML model should have a name and algorithm.

```python
class MLModel:

    def __init__(self, name, algorithm):
        self.name = name
        self.algorithm = algorithm


model = MLModel(
    "Fraud Detection Model",
    "XGBoost"
)

print(model.name)
print(model.algorithm)
```

4.Instance Variables

4.1 Understand the Concept

Instance variables are variables that belong to a specific object.
    
Each object can have different values for its instance variables.
    
They are usually created using self.
    
Example / Demonstration :
    
Suppose different ML models have different names and accuracy values.

```python
class MLModel:

    def __init__(self, name, accuracy):
        self.name = name
        self.accuracy = accuracy


model1 = MLModel("Fraud Model", 0.91)
model2 = MLModel("Customer Churn Model", 0.87)

print(model1.name)
print(model1.accuracy)

print(model2.name)
print(model2.accuracy)
```

5.Class Variables :
    
5.1 Understand the Concept

A class variable belongs to the class itself rather than to a particular object.
    
The same class variable is shared by all objects unless an object overrides it.
    
Example / Demonstration :

Suppose every model belongs to the same company.

```python
class MLModel:

    company = "AI Solutions"

    def __init__(self, name):
        self.name = name


model1 = MLModel("Fraud Model")
model2 = MLModel("Recommendation Model")

print(model1.company)
print(model2.company)
```

6.Instance Methods :

   
6.1 Understand the Concept
   
Instance methods are methods that operate on a specific object.

They use self as their first parameter and can access instance variables.

Example / Demonstration : 

Suppose an ML model has a method that displays its details.

```python
class MLModel:

    def __init__(self, name, accuracy):
        self.name = name
        self.accuracy = accuracy

    def display_details(self):
        print("Model:", self.name)
        print("Accuracy:", self.accuracy)


model = MLModel("Fraud Detection", 0.92)

model.display_details()
```

7.Static Methods :

7.1 Understand the Concept
   
A static method is a method that does not depend on a particular object or class.

It is created using the @staticmethod decorator.

Static methods are useful for utility operations related to a class.

Example / Demonstration :

Suppose we need a utility method to validate whether an accuracy value is between 0 and 1.

```python
class MLModel:

    @staticmethod
    def validate_accuracy(accuracy):
        return 0 <= accuracy <= 1


print(MLModel.validate_accuracy(0.92))
print(MLModel.validate_accuracy(1.5))
```

8.Class Methods :

8.1 Understand the Concept
   
A class method operates on the class rather than a specific object.

It is created using the @classmethod decorator.

The first parameter is usually cls.

Example / Demonstration

Suppose we want to keep track of how many ML model objects have been created.

```python
class MLModel:

    model_count = 0

    def __init__(self, name):
        self.name = name
        MLModel.model_count += 1

    @classmethod
    def get_model_count(cls):
        return cls.model_count


model1 = MLModel("Fraud Model")
model2 = MLModel("Recommendation Model")

print(MLModel.get_model_count())
```

9.Properties :
    
9.1 Understand the Concept

The @property decorator allows a method to be accessed like an attribute.

It is useful when we want to control how an attribute is accessed or calculated.

Example / Demonstration :

Suppose we store model accuracy as a decimal but want to display it as a percentage.

```python
class MLModel:

    def __init__(self, accuracy):
        self.accuracy = accuracy

    @property
    def accuracy_percentage(self):
        return self.accuracy * 100


model = MLModel(0.92)

print(model.accuracy_percentage)
```

10.Encapsulation :
    
10.1 Understand the Concept

Encapsulation means keeping data and the methods that operate on that data together inside a class.

It also allows us to control how internal data is accessed or modified.

Python commonly uses _ and __ naming conventions to indicate protected or private implementation details.

Example / Demonstration :

Suppose a model stores an internal prediction score that should be accessed through a method.

```python
class MLModel:

    def __init__(self, score):
        self.__score = score

    def get_score(self):
        return self.__score


model = MLModel(0.92)

print(model.get_score())
```

11.Abstraction:

11.1 Understand the Concept

Abstraction means hiding unnecessary implementation details and exposing only the functionality that users need.

In AI/ML systems, a user may call a predict() method without needing to know all the internal steps performed by the model.

Example / Demonstration :

Suppose different ML models must provide a prediction method.

```python
from abc import ABC, abstractmethod


class MLModel(ABC):

    @abstractmethod
    def predict(self, data):
        pass


class FraudModel(MLModel):

    def predict(self, data):
        return "Fraud prediction generated"


model = FraudModel()

print(model.predict([10000, 25000]))
```

12.Inheritance :

12.1 Understand the Concept

Inheritance allows one class to reuse properties and methods from another class.

The existing class is called the parent class, while the new class is called the child class.

Inheritance helps reduce duplicate code.
Example / Demonstration
Suppose we have a general ML model and a specific fraud detection model.

```python
class MLModel:

    def train(self):
        print("Model training started")


class FraudModel(MLModel):

    def detect_fraud(self):
        print("Fraud detection started")


model = FraudModel()

model.train()
model.detect_fraud()
```

13.Multiple Inheritance :

13.1 Understand the Concept

Multiple inheritance allows a class to inherit from more than one parent class.

This allows a class to combine functionality from multiple classes.

Example / Demonstration :

Suppose an AI system has separate classes for prediction and reporting.

```python
class Predictor:

    def predict(self):
        print("Generating prediction")


class Reporter:

    def generate_report(self):
        print("Generating model report")


class AIModel(Predictor, Reporter):
    pass


model = AIModel()

model.predict()
model.generate_report()
```

14.Multilevel Inheritance :

14.1 Understand the Concept

Multilevel inheritance occurs when a class inherits from another class that itself inherits from another class.

This creates a chain of inheritance.

Example / Demonstration :

Suppose we have a general system, an AI system, and a specific ML model.

```python
class System:

    def start(self):
        print("System started")


class AISystem(System):

    def process_data(self):
        print("AI system processing data")


class MLModel(AISystem):

    def predict(self):
        print("ML model generating prediction")


model = MLModel()

model.start()
model.process_data()
model.predict()
```

15.Hierarchical Inheritance :

15.1 Understand the Concept

Hierarchical inheritance occurs when multiple child classes inherit from the same parent class.

Example / Demonstration :

Suppose different AI models share common model functionality.

```python
class MLModel:

    def train(self):
        print("Training model")


class FraudModel(MLModel):

    def detect_fraud(self):
        print("Detecting fraud")


class RecommendationModel(MLModel):

    def recommend(self):
        print("Generating recommendation")


fraud_model = FraudModel()
recommendation_model = RecommendationModel()

fraud_model.train()
fraud_model.detect_fraud()

recommendation_model.train()
recommendation_model.recommend()
```

16.Polymorphism :

16.1 Understand the Concept

Polymorphism means that the same method or interface can behave differently depending on the object using it.

It allows different classes to provide their own implementation of the same operation.

Example / Demonstration :

Suppose different AI models have their own predict() behavior.

```python
class FraudModel:

    def predict(self):
        print("Fraud prediction generated")


class RecommendationModel:

    def predict(self):
        print("Product recommendation generated")


models = [
    FraudModel(),
    RecommendationModel()
]

for model in models:
    model.predict()
```

17: Method Overriding :

Step 1 – Understand the Concept

What is Method Overriding?

Method Overriding is an Object-Oriented Programming concept in which a child class provides its own implementation of a method that is already defined in its parent class.

The child class uses the same method name as the parent class method but changes its behavior according to its own requirements.

Method overriding allows a child class to inherit common functionality from the parent class while customizing specific behaviors.

In Python, when a method is called using an object of the child class, Python executes the overridden child method instead of the parent class method.

Step 2 – Demonstrate the Concept

Real-World Analogy

Consider an Animal system.

Every animal can make a sound, but different animals make different sounds.
An Animal can have a general sound() method.
A Dog can override sound() to bark.
A Cat can override sound() to meow.
 
Although all classes use the same method name, each child class provides its own implementation.

This demonstrates how a child class can customize inherited behavior.

Step 3 – Implementation Explanation

What does this program do?
This program defines a sound() method in the parent Animal class and then overrides the same method in the child Dog class.

When the method is called using the Dog object, Python executes the child's implementation.

Which part of the code demonstrates the concept?
The Dog class defines a sound() method with the same name as the method in Animal.
This demonstrates method overriding.

How does the concept work in this implementation?

When d.sound() is executed, Python checks the Dog class first.
Since Dog contains its own sound() method, Python executes that method instead of the inherited Animal.sound() method.

```python
class Animal:
    def sound(self):
        print("Animal makes a sound")


class Dog(Animal):
    def sound(self):
        print("Dog Barks")


d = Dog()
d.sound()
```

Topic 18: Method Overloading (Python Approach) :

Step 1 – Understand the Concept

What is Method Overloading?

Method Overloading is an OOP concept where multiple methods have the same name but accept different numbers or types of parameters.

we define multiple methods with the same name in Python, the latest definition replaces the previous one.

Instead, Python commonly achieves similar behavior using:

Default arguments
*args
**kwargs

This allows one method to handle different numbers or types of inputs.

Step 2 – Demonstrate the Concept

Real-World Analogy

Consider a Calculator.

A calculator may need to add:

Two numbers
Three numbers
Multiple numbers

Instead of creating completely different method names such as add_two(), add_three(), and add_many(), we can create one add() method that accepts different numbers of arguments.

Python can achieve this using *args.

Step 3 – Implementation Explanation

What does this program do?

This program creates an add() method that can accept any number of values.

The method uses *args to collect all the provided arguments and calculates their sum.

Which part of the code demonstrates the concept?

When different numbers of values are passed to add(), Python stores them inside args.

The method then uses sum() to calculate the total.



```python
class Calculator:

    def add(self, *args):
        return sum(args)


calc = Calculator()

print(calc.add(10, 20))
print(calc.add(10, 20, 30))
print(calc.add(10, 20, 30, 40))
```

Topic 19: Composition :

Step 1 – Understand the Concept

What is Composition?

Composition is an Object-Oriented Programming relationship in which one class contains an object of another class and strongly owns that object.

In composition, the contained object is strongly dependent on the main object. When the main object is destroyed, the contained object is generally no longer useful as part of that relationship.

Composition is useful when one object is considered an essential part of another object.

Step 2 – Demonstrate the Concept

Real-World Analogy

Consider a Car and an Engine.

A car has an engine.

The engine is created as part of the car in this example.

Car is the main object.
Engine is the component.
Car creates and owns the Engine.

This represents a strong has-a relationship.

Step 3 – Implementation Explanation

What does this program do?

This program creates an Engine class and a Car class.
The Car class creates an Engine object inside its constructor.

```python
class Engine:
    def start(self):
        print("Engine started")


class Car:
    def __init__(self):
        self.engine = Engine()

    def start_car(self):
        self.engine.start()
        print("Car started")


car = Car()
car.start_car()
```

Topic 20: Aggregation :

Step 1 – Understand the Concept

What is Aggregation?

Aggregation is an Object-Oriented Programming relationship where one class contains or uses objects of another class, but the contained objects can exist independently.

Aggregation represents a weaker "has-a" relationship compared with composition.
The child or contained object has its own independent existence.

For example, a Department can have Employee objects, but an employee can exist even if the department is removed.

Step 2 – Demonstrate the Concept

Real-World Analogy

Consider a Department and an Employee.

A department has employees.

However, an employee does not depend completely on the department for existence.
An employee can leave one department and join another department.

Therefore, this represents aggregation.

Step 3 – Implementation Explanation

What does this program do?

This program creates Employee objects separately and then passes them into a Department.

```python
class Employee:
    def __init__(self, name):
        self.name = name

    def display(self):
        print("Employee:", self.name)


class Department:
    def __init__(self, employees):
        self.employees = employees

    def show_employees(self):
        for employee in self.employees:
            employee.display()


employee1 = Employee("harsha")
employee2 = Employee("Prateek")

department = Department([employee1, employee2])

department.show_employees()
```

What is Association?

Association is a relationship between two independent classes where objects of one class interact with objects of another class.

Association represents a general relationship between objects.

Unlike composition and aggregation, association does not necessarily mean that one object owns another object.

It simply indicates that two objects know about, communicate with, or work with each other.

Consider a Teacher and a Student.

A teacher teaches a student.

However:

Topic 22: Magic (Dunder) Methods :

Step 1 – Understand the Concept

What are Magic (Dunder) Methods?

Magic methods, also called dunder methods, are special methods in Python whose names begin and end with double underscores.

"Dunder" means double underscore.

These methods allow Python objects to interact with built-in Python operations and functions.

For example, when we use len(object), Python internally looks for the object's __len__() method.

Similarly, when we use print(object), Python can use the __str__() method.

Magic methods allow classes to define how their objects behave with Python's built-in operations.

Step 2 – Demonstrate the Concept 

Real-World Analogy

Consider a Shopping Cart.
    
A shopping cart may need to:
Display itself when printed.
Tell how many products it contains.
Represent itself for debugging.

Python can provide these behaviors through special methods such as:
__str__()
__len__()
__repr__()

These methods allow normal Python operations to work naturally with custom objects.
Step 3 – Implementation Explanation

What does this program do?

This program defines a ShoppingCart class and implements special methods to control how the object behaves.

```python
class ShoppingCart:

    def __init__(self, items):
        self.items = items

    def __str__(self):
        return "Shopping Cart"

    def __len__(self):
        return len(self.items)

    def __repr__(self):
        return f"ShoppingCart({self.items!r})"


cart = ShoppingCart(["Laptop", "Mouse"])

print(cart)
print(len(cart))
print(repr(cart))
```

Topic 23: super() 

Step 1 – Understand the Concept

What is super()?
super() is a built-in Python function used to access methods and attributes from a parent class.

It is commonly used when working with inheritance.

The most common use of super() is inside a child class constructor to call the parent class's __init__() method.

super() allows a child class to reuse functionality from its parent class instead of duplicating the same code.

Step 2 – Demonstrate the Concept

Real-World Analogy

Consider an Employee and a Manager.
An employee has a name.
A manager is also an employee but has an additional department.
Instead of writing the employee initialization again inside Manager, we can use super() to call the parent constructor.
The child class can then add its own additional information.

Step 3 – Implementation Explanation

What does this program do?

This program defines an Employee parent class and a Manager child class.
The Manager uses super() to call the constructor of Employee.

```python
class Employee:
    def __init__(self, name):
        self.name = name


class Manager(Employee):
    def __init__(self, name, department):
        super().__init__(name)
        self.department = department

    def display(self):
        print("Name:", self.name)
        print("Department:", self.department)


manager = Manager("prateek", "AI/ML")
manager.display()
```

Topic 24: __str__()

Step 1 – Understand the Concept

What is __str__()?

__str__() is a special Python method that defines the human-readable string representation of an object.

It is automatically called when an object is passed to print() or when str() is used.

Without defining __str__(), Python normally displays a default representation containing information about the object's class and memory location.

By implementing __str__(), we can make the output easier for users to understand.

```python
class Student:
    def __init__(self, name, course):
        self.name = name
        self.course = course

    def __str__(self):
        return f"Student Name: {self.name}, Course: {self.course}"


student = Student("prateek", "AI/ML")

print(student)
```

Topic 25: __repr__()

Step 1 – Understand the Concept

What is __repr__()?

__repr__() is a special Python method that defines the developer-oriented representation of an object.

It is mainly useful for debugging and development.

While __str__() focuses on a readable representation for users, __repr__() aims to provide a more detailed and unambiguous representation of the object.

When repr(object) is called, Python uses the object's __repr__() method.

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def __str__(self):
        return self.name

    def __repr__(self):
        return f"Product(name={self.name!r}, price={self.price!r})"


product = Product("Laptop", 75000)

print(product)
print(repr(product))
```

Topic 26: __len__()

Step 1 – Understand the Concept

What is __len__()?

__len__() is a special Python method used to define what should happen when the built-in len() function is used with an object.

For normal Python collections such as lists and strings, len() returns the number of elements or characters.

By implementing __len__() in a custom class, we can define what the length of our object means.

```python
class Playlist:
    def __init__(self, songs):
        self.songs = songs

    def __len__(self):
        return len(self.songs)


playlist = Playlist(["Song A", "Song B", "Song C"])

print(len(playlist))
```

Topic 28: Singleton Pattern (Bonus) :

Step 1 – Understand the Concept

What is the Singleton Pattern?

The Singleton Pattern is a design pattern that ensures a class has only one instance during the lifetime of an application.

Instead of creating a new object every time the class is called, the Singleton pattern returns the same object.

It can be useful when an application needs one shared instance of something such as a configuration manager or application-wide service.

Step 2 – Demonstrate the Concept

Real-World Analogy

Consider an application's Configuration Manager.
The application may need one central configuration object.
Instead of creating multiple configuration manager objects, the Singleton pattern ensures that different parts of the application use the same instance.

Step 3 – Implementation Explanation

What does this program do?

This program creates a Singleton class that maintains only one instance of itself.
When the class is called multiple times, the same object is returned.
Which part of the code demonstrates the concept?

The __new__() method checks whether an instance already exists.

If it does not exist, a new instance is created.

If it already exists, the existing instance is returned.
How does the concept work in this implementation?

The class variable _instance stores the single object.
The first time Singleton() is called, an object is created.

```python
class Singleton:

    _instance = None

    def __new__(cls):
        if cls._instance is None:
            cls._instance = super().__new__(cls)

        return cls._instance


object1 = Singleton()
object2 = Singleton()

print(object1 is object2)
```
