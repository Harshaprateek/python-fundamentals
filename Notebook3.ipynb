Notebook 3 – Exception Handling

Topic 1: try

Step 1 – Understand the Concept

What is try?

The try block in Python is used to contain code that may produce an exception during program execution.

An exception is an error that occurs while a program is running. Instead of allowing the program to terminate unexpectedly, Python allows us to handle such errors using exception handling.

The try block contains the code that we want Python to monitor for possible exceptions.

The try block is normally used together with except, else, and finally.

```python
try:
    number = 10 / 2
    print(number)
except ZeroDivisionError:
    print("Cannot divide by zero")
```

Topic 2: except

Step 1 – Understand the Concept

What is except?

The except block is used to handle an exception raised inside a try block.

When an exception occurs, Python stops executing the remaining statements in the try block and looks for a matching except block.

This allows the program to handle errors without terminating unexpectedly.

Step 2 – Demonstrate the Concept
Real-World Analogy

If an ATM transaction fails, the system displays an appropriate error message instead of crashing.

The error-handling action is similar to an except block.

```python
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Cannot divide by zero")
```

Topic 3: else

Step 1 – Understand the Concept

What is else in Exception Handling?

The else block is executed only when the code inside the try block completes successfully without raising an exception.

It is useful when we want to separate successful execution from error-handling code.

Step 2 – Demonstrate the Concept
Real-World Analogy

Consider making an online payment.

If the payment succeeds, the application displays a success message.

If the payment fails, it handles the error.

The successful action can be represented using else.

```python
try:
    result = 20 / 4
except ZeroDivisionError:
    print("Cannot divide by zero")
else:
    print("Division successful")
    print("Result:", result)
```

Topic 4: finally

Step 1 – Understand the Concept

What is finally?

The finally block contains code that should execute regardless of whether an exception occurs.

It is commonly used for cleanup operations such as:

Closing files
Closing database connections
Releasing resources
Closing network connections

The finally block executes whether the try block succeeds or fails.

Step 2 – Demonstrate the Concept
Real-World Analogy

When you rent a car, whether the journey is successful or not, you eventually return the car.

The cleanup action is similar to what happens inside finally.

```python
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Cannot divide by zero")
finally:
    print("Execution completed")
```

Topic 5: raise

Step 1 – Understand the Concept

What is raise?

The raise statement is used to manually generate an exception.

Instead of waiting for Python to automatically generate an error, a programmer can use raise when a specific condition is invalid.

This is useful for validating input and enforcing application rules.

Step 2 – Demonstrate the Concept
Real-World Analogy

A bank may reject a withdrawal when the amount is greater than the available balance.

The bank system intentionally raises an error because the requested operation violates a rule.

```python
age = -5

try:
    if age < 0:
        raise ValueError("Age cannot be negative")

    print("Valid age")

except ValueError as error:
    print("Error:", error)
```

Topic 6: assert

Step 1 – Understand the Concept

What is assert?

assert is used to test whether a condition is true.

If the condition is true, the program continues normally.

If the condition is false, Python raises an AssertionError.

Assertions are commonly used during development and testing to verify assumptions about program state.

Step 2 – Demonstrate the Concept
Real-World Analogy

Before allowing a user to enter a restricted area, a security system may check whether the person has a valid ID.

If the condition is not satisfied, access is rejected.

```python
number = -10

try:
    assert number > 0, "Number must be positive"
    print("Valid number")
except AssertionError as error:
    print("Assertion failed:", error)
```

Topic 7: Custom Exceptions

Step 1 – Understand the Concept

What are Custom Exceptions?

A custom exception is a user-defined exception created for a specific application requirement.

Python provides many built-in exceptions, but sometimes an application needs a more meaningful error type.

Custom exceptions are created by inheriting from the built-in Exception class.

Step 2 – Demonstrate the Concept
Real-World Analogy

Consider a banking application.

```python
class InsufficientBalanceError(Exception):
    pass


balance = 5000
withdrawal = 7000

try:
    if withdrawal > balance:
        raise InsufficientBalanceError("Insufficient balance")

    balance -= withdrawal
    print("Withdrawal successful")

except InsufficientBalanceError as error:
    print("Error:", error)
```

Topic 8: ValueError

Step 1 – Understand the Concept

What is ValueError?

ValueError occurs when a function receives a value of the correct type but the value itself is inappropriate or invalid.

For example, converting the string "hello" into an integer causes a ValueError.

Step 2 – Demonstrate the Concept
Real-World Analogy

A form may require a person's age as a number.

```python
try:
    age = int("hello")
    print(age)

except ValueError:
    print("Please enter a valid number")
```

Topic 9: TypeError

Step 1 – Understand the Concept

What is TypeError?

TypeError occurs when an operation is performed on an object of an inappropriate type.

For example, adding a string and an integer using + causes a TypeError.

Step 2 – Demonstrate the Concept
Real-World Analogy

You cannot directly add a physical book and a number.

```python
try:
    result = "10" + 5
    print(result)

except TypeError:
    print("Cannot add a string and an integer directly")
```

Topic 10: IndexError

Step 1 – Understand the Concept

What is IndexError?

IndexError occurs when we try to access an index that does not exist in a sequence such as a list or tuple.

Python uses zero-based indexing.

Step 2 – Demonstrate the Concept
Real-World Analogy

If a shelf contains three books, trying to access the fifth book causes a problem because that position does not exist.

```python
numbers = [10, 20, 30]

try:
    print(numbers[5])

except IndexError:
    print("Index is outside the list range")
```

Topic 11: KeyError

Step 1 – Understand the Concept

What is KeyError?

KeyError occurs when we try to access a dictionary using a key that does not exist.

Dictionaries use keys to access their values.

Step 2 – Demonstrate the Concept
Real-World Analogy

If an employee database contains an employee's name and department but does not contain their salary, requesting the missing salary field causes a proble

```python
employee = {
    "name": "Rahul",
    "department": "AI/ML"
}

try:
    print(employee["salary"])

except KeyError:
    print("The requested key does not exist")
```

Topic 12: AttributeError

Step 1 – Understand the Concept

What is AttributeError?

AttributeError occurs when an object does not have the attribute or method that we try to access.

Step 2 – Demonstrate the Concept
Real-World Analogy

If an employee record contains name and department, asking the system for a non-existent salary attribute causes an error.

```python
class Employee:
    def __init__(self, name):
        self.name = name


employee = Employee("Rahul")

try:
    print(employee.salary)

except AttributeError:
    print("The object does not have this attribute")
```

Topic 14: FileNotFoundError

Step 1 – Understand the Concept

What is FileNotFoundError?

FileNotFoundError occurs when Python tries to access a file that does not exist at the specified location.

It commonly occurs while reading files.

Step 2 – Demonstrate the Concept
Real-World Analogy

```python
try:
    with open("missing_file.txt", "r") as file:
        content = file.read()

except FileNotFoundError:
    print("The requested file was not found")
```

Topic 16: ImportError

Step 1 – Understand the Concept

What is ImportError?

ImportError occurs when Python is unable to import a requested object from a module.

The module itself may exist, but the specific item being imported may not exist.

Step 2 – Demonstrate the Concept
Real-World Analogy

A library may exist, but the specific book you request may not be available in that library.

```python
try:
    from math import unknown_function

except ImportError:
    print("The requested item could not be imported")
```
