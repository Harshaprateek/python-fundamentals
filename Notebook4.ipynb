Topic 10: Log Formatters
Step 1 – Understand the Concept
What are Log Formatters?

A formatter can include information such as:

Time
Log level
Logger name
Message
Step 2 – Demonstrate the Concept
Real-World Analogy

An official report usually follows a fixed format containing the date, category, and description.

```python
import logging

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)

logging.info("Application started")
logging.warning("Low memory")
```

Topic 11: File Handlers
Step 1 – Understand the Concept
What are File Handlers?

A FileHandler sends log messages to a file instead of displaying them only on the console.

This allows applications to maintain persistent log records.

Step 2 – Demonstrate the Concept
Real-World Analogy

Instead of announcing an event verbally, an organization writes the event into an official logbook.

```python
import logging

logger = logging.getLogger("application")
logger.setLevel(logging.INFO)

file_handler = logging.FileHandler("application.log")

formatter = logging.Formatter(
    "%(asctime)s - %(levelname)s - %(message)s"
)

file_handler.setFormatter(formatter)

logger.addHandler(file_handler)

logger.info("Application started")
logger.error("Example error")
```

Topic 12: Console Handlers
Step 1 – Understand the Concept
What are Console Handlers?

A StreamHandler can be used to send logging messages to the console.

This is useful when developers want to see application activity directly while the program is running.

Step 2 – Demonstrate the Concept

A developer running an application can see important events directly in the terminal.

```python
import logging

logger = logging.getLogger("application")
logger.setLevel(logging.INFO)

console_handler = logging.StreamHandler()

formatter = logging.Formatter(
    "%(levelname)s - %(message)s"
)

console_handler.setFormatter(formatter)

logger.addHandler(console_handler)

logger.info("Application running")
```

Topic 13: Rotating Logs (Bonus)
Step 1 – Understand the Concept
What are Rotating Logs?

Log files can become very large if an application runs for a long time.

Rotating logs automatically create new log files when the existing file reaches a specified size or time limit.

```python
import logging
from logging.handlers import RotatingFileHandler

logger = logging.getLogger("application")
logger.setLevel(logging.INFO)

handler = RotatingFileHandler(
    "application.log",
    maxBytes=1000,
    backupCount=3
)

formatter = logging.Formatter(
    "%(asctime)s - %(levelname)s - %(message)s"
)

handler.setFormatter(formatter)
logger.addHandler(handler)

for i in range(100):
    logger.info(f"Log message {i}")
```

Topic 10: File Validation
Step 1 – Understand the Concept
What is File Validation?

File validation means checking whether a file meets certain requirements before using it.

Validation can check:

Whether the file exists
File extension
File type
File size
Whether the file can be opened

File validation helps prevent errors when processing files.

Step 2 – Demonstrate the Concept
Real-World Analogy

Before accepting a document, an application may check whether it is a valid PDF and whether its size is within the allowed limit.

```python
from pathlib import Path

file_path = Path("sample.txt")

if not file_path.exists():
    print("File does not exist")

elif file_path.suffix != ".txt":
    print("Invalid file type")

elif file_path.stat().st_size == 0:
    print("File is empty")

else:
    print("File is valid")
```

Notebook 5 – Final File Handling Example
Step 3 – Implementation Explanation
What does this program do?

This final example combines several file-handling concepts.

It creates a directory, creates a text file inside it, writes information, reads the information, and checks whether the file exists.

```python
from pathlib import Path

folder = Path("file_demo")
folder.mkdir(exist_ok=True)

file_path = folder / "example.txt"

file_path.write_text("Python File Handling")

with open(file_path, "r") as file:
    content = file.read()

print("File exists:", file_path.exists())
print("File content:", content)
```
