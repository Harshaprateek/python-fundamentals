1. Database Design
Step 1 – Understand the Concept
What is Database Design?

Database design is the process of organizing data into tables and defining how those tables are related.

A good database design helps us:

Store data efficiently
Avoid duplicate data
Maintain data accuracy
Create relationships between tables
Retrieve data easily

A typical company database may contain:

Employees
Departments
Projects
Customers
Orders
Step 2 – Demonstrate the Concept
Real-World Analogy

Think about a company.

Instead of storing everything in one huge table:

Employee	Department	Manager
Rahul	IT	Amit
Priya	HR	Sneha

we can separate the information:

Departments

department_id	department_name
1	IT
2	HR

Employees

employee_id	name	department_id
101	prateek	1
102	harsha	2

The department_id connects the tables.

```python
import sqlite3

connection = sqlite3.connect("company.db")
cursor = connection.cursor()

print("SQLite database connected successfully")
```

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS Departments (
    department_id INTEGER PRIMARY KEY,
    department_name TEXT NOT NULL
)
""")

connection.commit()

print("Departments table created")
```

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS Employees (
    employee_id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    salary REAL,
    department_id INTEGER,
    FOREIGN KEY (department_id)
        REFERENCES Departments(department_id)
)
""")

connection.commit()

print("Employees table created")
```

19. CROSS JOIN
Step 1 – Understand the Concept

CROSS JOIN produces every possible combination of rows from two tables.

Step 2 – Demonstrate the Concept

If there are:

3 employees
2 colors

the result contains:

3 × 2 = 6 combinations


```python
import sqlite3

# -----------------------------------
# DATABASE CONNECTION
# -----------------------------------

connection = sqlite3.connect("company.db")
cursor = connection.cursor()

print("Database connected successfully")


# -----------------------------------
# CREATE TABLES
# -----------------------------------

cursor.execute("""
CREATE TABLE IF NOT EXISTS Departments (
    department_id INTEGER PRIMARY KEY,
    department_name TEXT NOT NULL
)
""")

cursor.execute("""
CREATE TABLE IF NOT EXISTS Employees (
    employee_id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    salary REAL,
    department_id INTEGER
)
""")

cursor.execute("""
CREATE TABLE IF NOT EXISTS Colors (
    color TEXT
)
""")

connection.commit()


# -----------------------------------
# CLEAR OLD DEMO DATA
# -----------------------------------

cursor.execute("DELETE FROM Employees")
cursor.execute("DELETE FROM Departments")
cursor.execute("DELETE FROM Colors")

connection.commit()


# -----------------------------------
# INSERT DEPARTMENTS
# -----------------------------------

cursor.execute("""
INSERT INTO Departments
VALUES (1, 'IT')
""")

cursor.execute("""
INSERT INTO Departments
VALUES (2, 'HR')
""")

cursor.execute("""
INSERT INTO Departments
VALUES (3, 'Finance')
""")

cursor.execute("""
INSERT INTO Departments
VALUES (4, 'Marketing')
""")

connection.commit()


# -----------------------------------
# INSERT EMPLOYEES
# -----------------------------------

cursor.execute("""
INSERT INTO Employees
VALUES (101, 'Rahul', 75000, 1)
""")

cursor.execute("""
INSERT INTO Employees
VALUES (102, 'Priya', 65000, 2)
""")

cursor.execute("""
INSERT INTO Employees
VALUES (103, 'Amit', 85000, 1)
""")

# Employee without a department
cursor.execute("""
INSERT INTO Employees
VALUES (104, 'Kiran', 55000, NULL)
""")

connection.commit()


# -----------------------------------
# INSERT COLORS
# -----------------------------------

cursor.execute("""
INSERT INTO Colors
VALUES ('Red')
""")

cursor.execute("""
INSERT INTO Colors
VALUES ('Blue')
""")

connection.commit()


# ==========================================================
# 14. LIMIT
# ==========================================================

print("\n14. LIMIT")
print("--------------------")

cursor.execute("""
SELECT *
FROM Employees
LIMIT 2
""")

for row in cursor.fetchall():
    print(row)


# ==========================================================
# 15. INNER JOIN
# ==========================================================

print("\n15. INNER JOIN")
print("--------------------")

cursor.execute("""
SELECT
    Employees.name,
    Employees.salary,
    Departments.department_name
FROM Employees
INNER JOIN Departments
ON Employees.department_id = Departments.department_id
""")

for row in cursor.fetchall():
    print(row)


# ==========================================================
# 16. LEFT JOIN
# ==========================================================

print("\n16. LEFT JOIN")
print("--------------------")

cursor.execute("""
SELECT
    Departments.department_name,
    Employees.name
FROM Departments
LEFT JOIN Employees
ON Departments.department_id = Employees.department_id
""")

for row in cursor.fetchall():
    print(row)


# ==========================================================
# 17. RIGHT JOIN
# ==========================================================

print("\n17. RIGHT JOIN")
print("--------------------")

cursor.execute("""
SELECT
    Departments.department_name,
    Employees.name
FROM Departments
RIGHT JOIN Employees
ON Departments.department_id = Employees.department_id
""")

for row in cursor.fetchall():
    print(row)


# ==========================================================
# 18. FULL JOIN
# ==========================================================

print("\n18. FULL JOIN")
print("--------------------")

cursor.execute("""
SELECT
    Departments.department_name,
    Employees.name
FROM Departments
FULL OUTER JOIN Employees
ON Departments.department_id = Employees.department_id
""")

for row in cursor.fetchall():
    print(row)


# ==========================================================
# 19. CROSS JOIN
# ==========================================================

print("\n19. CROSS JOIN")
print("--------------------")

cursor.execute("""
SELECT
    Employees.name,
    Colors.color
FROM Employees
CROSS JOIN Colors
""")

for row in cursor.fetchall():
    print(row)

```

21. CASE
Step 1 – Understand the Concept

CASE works like an if-elif-else statement in SQL.

Step 2 – Demonstrate the Concept

Classify employees based on salary.

>= 80000 → High
>= 60000 → Medium
otherwise → Low

```python
cursor.execute("""
SELECT
    name,
    salary,
    CASE
        WHEN salary >= 80000 THEN 'High'
        WHEN salary >= 60000 THEN 'Medium'
        ELSE 'Low'
    END AS salary_level
FROM Employees
""")

for row in cursor.fetchall():
    print(row)
```

22. Subqueries
Step 1 – Understand the Concept

A subquery is a query inside another query.

Step 2 – Demonstrate the Concept

Find employees earning more than the average salary.

First calculate:

Average salary

Then find employees whose salary is greater than that value.

```python
cursor.execute("""
SELECT
    name,
    salary
FROM Employees
WHERE salary > (
    SELECT AVG(salary)
    FROM Employees
)
""")

for row in cursor.fetchall():
    print(row)
```

23. Common Table Expressions – CTE
Step 1 – Understand the Concept

A CTE allows us to temporarily name the result of a query using WITH.

It makes complex queries easier to read.

Step 2 – Demonstrate the Concept

First calculate average salary by department.

Then use that temporary result.

```python
cursor.execute("""
WITH DepartmentSalary AS (
    SELECT
        department_id,
        AVG(salary) AS average_salary
    FROM Employees
    GROUP BY department_id
)
SELECT *
FROM DepartmentSalary
""")

for row in cursor.fetchall():
    print(row)
```

26. Window Functions
Step 1 – Understand the Concept

Window functions perform calculations across related rows without combining those rows into one result.

They are useful for:

Ranking
Previous values
Next values
Running calculations
Comparisons between rows
27. ROW_NUMBER()
Step 1 – Understand the Concept

ROW_NUMBER() assigns a unique sequential number to each row.

Step 2 – Demonstrate the Concept

Rank employees according to salary.

```python
cursor.execute("""
SELECT
    name,
    salary,
    ROW_NUMBER() OVER (
        ORDER BY salary DESC
    ) AS row_number
FROM Employees
""")

for row in cursor.fetchall():
    print(row)
```

28. RANK()
Step 1 – Understand the Concept

RANK() assigns the same rank to rows having the same value.

Ties create gaps in ranking.

Example:

100 → 1
100 → 1
90  → 3
Step 2 – Demonstrate the Concept

Rank employees based on salary.

```python
cursor.execute("""
SELECT
    name,
    salary,
    RANK() OVER (
        ORDER BY salary DESC
    ) AS salary_rank
FROM Employees
""")

for row in cursor.fetchall():
    print(row)
```

29. DENSE_RANK()
Step 1 – Understand the Concept

DENSE_RANK() is similar to RANK() but does not leave gaps.

Example:

100 → 1
100 → 1
90  → 2
Step 2 – Demonstrate the Concept

Rank employees by salary without ranking gaps.

```python
cursor.execute("""
SELECT
    name,
    salary,
    DENSE_RANK() OVER (
        ORDER BY salary DESC
    ) AS salary_rank
FROM Employees
""")

for row in cursor.fetchall():
    print(row)
```

30. LAG()
Step 1 – Understand the Concept

LAG() accesses a previous row.

It is useful for comparing the current value with the previous value.

Step 2 – Demonstrate the Concept

For monthly sales:

January → previous = NULL
February → previous = January
March → previous = February

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS MonthlySales (
    month_number INTEGER PRIMARY KEY,
    month_name TEXT,
    sales REAL
)
""")

connection.commit()
```

```python
cursor.execute("DELETE FROM MonthlySales")

cursor.execute("""
INSERT INTO MonthlySales VALUES
(1, 'January', 50000)
""")

cursor.execute("""
INSERT INTO MonthlySales VALUES
(2, 'February', 55000)
""")

cursor.execute("""
INSERT INTO MonthlySales VALUES
(3, 'March', 62000)
""")

cursor.execute("""
INSERT INTO MonthlySales VALUES
(4, 'April', 58000)
""")

connection.commit()
```

```python
cursor.execute("""
SELECT
    month_name,
    sales,
    LAG(sales) OVER (
        ORDER BY month_number
    ) AS previous_month_sales
FROM MonthlySales
""")

for row in cursor.fetchall():
    print(row)
```

31. LEAD()
Step 1 – Understand the Concept

LEAD() accesses the value from the next row.

Step 2 – Demonstrate the Concept

For monthly sales:

January → next = February
February → next = March
March → next = April

```python
cursor.execute("""
SELECT
    month_name,
    sales,
    LEAD(sales) OVER (
        ORDER BY month_number
    ) AS next_month_sales
FROM MonthlySales
""")

for row in cursor.fetchall():
    print(row)
```

```python
cursor.execute("""
WITH DepartmentAverage AS (
    SELECT
        department_id,
        AVG(salary) AS average_salary
    FROM Employees
    GROUP BY department_id
)
SELECT
    E.name,
    E.salary,
    E.department_id,

    CASE
        WHEN E.salary >= 80000 THEN 'High'
        WHEN E.salary >= 60000 THEN 'Medium'
        ELSE 'Low'
    END AS salary_level,

    DENSE_RANK() OVER (
        PARTITION BY E.department_id
        ORDER BY E.salary DESC
    ) AS department_rank

FROM Employees E
JOIN DepartmentAverage D
ON E.department_id = D.department_id
""")

for row in cursor.fetchall():
    print(row)
```
