Notebook 1: Introduction to Statistics

1. What is Statistics?
   
Understand what statistics means and how we collect, organize, analyze and interpret data.

3. Why Statistics is Important in AI/ML
   
Understand how statistics helps with:

- Understanding datasets
- Finding patterns
- Data preprocessing
- Model evaluation
- Making decisions from data

3. Statistics vs Machine Learning
  
Understand the difference between:

- Statistics → understanding data and relationships
- Machine Learning → learning patterns from data to make predictions/decisions

4. Descriptive Statistics
Cover how we summarize existing data using things such as:
- Mean
- Median
- Mode
- Minimum
- Maximum
- Range
- Standard deviation
5. Inferential Statistics
Understand how we use a sample to draw conclusions about a larger population.
6. Population
The complete group we are interested in studying.
7. Sample
A smaller subset selected from the population.
8. Census
Collecting information from every member of the population.
9. Sampling
The process of selecting a sample from a population.
10. Variables
A characteristic or value that can change between observations.
11. Features
Input variables used by a machine-learning model to make predictions.
12. Observation
One individual record/data point in a dataset.
13. Dataset
A collection of observations and their variables/features.
14. Quantitative Data
Numerical data that can be measured or counted.
Examples:
- Age = 25
- Salary = ₹50,000
- Height = 176 cm
15. Qualitative Data
Descriptive or categorical data rather than numerical measurements.
Examples:
- Gender/category
- Product type
- City
- Customer feedback
16. Continuous Data
Numerical data that can take values within a range.
Examples:
- Height = 176.5 cm
- Temperature = 36.7°C
- Weight = 72.4 kg
17. Discrete Data
Countable numerical values, usually whole numbers.
Examples:
- Number of customers = 100
- Number of orders = 25
- Number of cars = 8
18. Structured Data
Data organized into a predefined structure, usually rows and columns.
Example:
Age	Salary	City
25	50000	Hyderabad
30	65000	Chennai


This is common in databases and CSV files.
19. Unstructured Data
Data without a fixed tabular structure.
Examples:
- Images
- Videos
- Audio
- Text documents
- Social-media posts
This is especially important in AI/ML, NLP, computer vision and Generative AI.
