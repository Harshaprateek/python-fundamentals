## 1. Null Hypothesis

### Definition

The Null Hypothesis, represented by H₀, is the default assumption that there is no significant effect, difference, or relationship.

### Example

A company claims that the average delivery time is 30 minutes.

We can define:

H₀: Average delivery time = 30 minutes

The null hypothesis represents the default assumption that the company's claim is true.

### Business Example

A company introduces a new website design.

H₀: The new design does not change the conversion rate.

### AI/ML Example

H₀ can be used when comparing model performance to determine whether an observed difference is statistically significant.

## 2. Alternative Hypothesis

### Definition

The Alternative Hypothesis, represented by H₁ or Hₐ, is the claim that there is a significant effect, difference, or relationship.

### Example

If:

H₀: Average delivery time = 30 minutes

Then:

Hₐ: Average delivery time ≠ 30 minutes

### Business Example

H₀: New marketing campaign does not increase sales.

Hₐ: New marketing campaign increases sales.

### Important

Hypothesis testing uses sample data to determine whether there is enough evidence to reject the null hypothesis.

## 3. p-value

### Definition

The p-value measures how compatible the observed data is with the null hypothesis.

A small p-value indicates that the observed result would be relatively unusual if the null hypothesis were true.

### Common Decision Rule

If:

p-value < significance level

→ Reject H₀

If:

p-value ≥ significance level

→ Fail to reject H₀

### Example

Suppose:

p-value = 0.02

Significance level = 0.05

Because:

0.02 < 0.05

We reject H₀.

### Important

A p-value is NOT the probability that the null hypothesis is true.

It is calculated under the assumption that the null hypothesis is true.

## 4. Confidence Interval

### Definition

A confidence interval provides a range of plausible values for a population parameter based on sample data.

A commonly used interval is the 95% confidence interval.

### Example

Suppose a sample produces:

Mean = 75

95% Confidence Interval = [72, 78]

This gives a range of plausible values for the population mean under the assumptions of the method used.

### Business Example

A company estimates that the average customer spending is between ₹950 and ₹1,050.

### AI/ML Example

Confidence intervals can be used to quantify uncertainty around statistical estimates and model metrics.

## 5. Significance Level

### Definition

The significance level, represented by α (alpha), is the threshold used to decide whether evidence is strong enough to reject the null hypothesis.

Common values:

- 0.10
- 0.05
- 0.01

The most commonly used significance level is 0.05.

### Example

α = 0.05

If p-value < 0.05:

Reject H₀.

If p-value ≥ 0.05:

Fail to reject H₀.

### Interpretation

A 0.05 significance level means we use a 5% threshold for the decision rule in the hypothesis test.

## 6. Type I Error

### Definition

A Type I Error occurs when we reject the null hypothesis even though it is actually true.

It is also called a false positive.

### Example

Suppose:

H₀: A customer is not fraudulent.

We reject H₀ and classify the customer as fraudulent.

But the customer was actually legitimate.

This is a Type I Error.

### Business Example

A fraud detection system incorrectly flags a legitimate transaction as fraudulent.

### Relationship with Significance Level

The significance level α controls the probability of a Type I Error under the assumptions of the test.

## 7. Type II Error

### Definition

A Type II Error occurs when we fail to reject the null hypothesis even though the alternative hypothesis is actually true.

It is also called a false negative.

### Example

Suppose:

H₀: A transaction is legitimate.

The transaction is actually fraudulent, but the system does not detect it.

This is a Type II Error.

### Business Example

A fraud detection model fails to detect a fraudulent transaction.

### Important Concept

Statistical power is:

Power = 1 - β

where β is the probability of a Type II Error.

## 8. One-Tailed Test

### Definition

A one-tailed test checks for an effect in one specific direction.

### Example

A company claims that a new process reduces average delivery time.

H₀: μ ≥ 30

Hₐ: μ < 30

The alternative hypothesis specifies a decrease.

### Types

#### Left-tailed

Tests whether the parameter is significantly smaller.

#### Right-tailed

Tests whether the parameter is significantly larger.

### Business Example

Testing whether a new process increases sales.

Hₐ: New sales > Old sales.

## 9. Two-Tailed Test

### Definition

A two-tailed test checks for a difference in either direction.

### Example

A company claims:

Average delivery time = 30 minutes.

We want to determine whether the actual average is different.

H₀: μ = 30

Hₐ: μ ≠ 30

The actual value could be:

- Greater than 30
- Less than 30

### Business Example

Testing whether a new system changes average processing time, without specifying whether it will increase or decrease it.

## Hypothesis Testing Example

Suppose a company claims that the average delivery time is 30 minutes.

We collect a sample of delivery times.

### Hypotheses

H₀: μ = 30

Hₐ: μ ≠ 30

### Significance Level

α = 0.05

We can perform a one-sample t-test to determine whether the sample provides sufficient evidence that the population mean differs from 30 minutes.

```python
import numpy as np
from scipy import stats

delivery_times = np.array([
    28, 31, 29, 32, 30,
    27, 33, 31, 29, 30
])

population_mean = 30

t_statistic, p_value = stats.ttest_1samp(
    delivery_times,
    population_mean
)

print("T-statistic:", t_statistic)
print("P-value:", p_value)

alpha = 0.05

if p_value < alpha:
    print("Reject the null hypothesis.")
else:
    print("Fail to reject the null hypothesis.")
```

### Interpretation

We compare the p-value with the significance level of 0.05.

If:

p-value < 0.05

we reject H₀.

Otherwise, we fail to reject H₀.

Failing to reject H₀ does not prove that H₀ is true. It means the sample does not provide sufficient evidence against H₀ at the chosen significance level.

```python
confidence_level = 0.95

sample_mean = np.mean(delivery_times)
sample_std = np.std(delivery_times, ddof=1)
sample_size = len(delivery_times)

standard_error = sample_std / np.sqrt(sample_size)

confidence_interval = stats.t.interval(
    confidence_level,
    df=sample_size - 1,
    loc=sample_mean,
    scale=standard_error
)

print("Sample Mean:", sample_mean)
print("95% Confidence Interval:", confidence_interval)
```

### Interpretation

The confidence interval gives a range of plausible values for the population mean based on this sample and the assumptions of the t-based interval.

A confidence interval provides information about both the estimated value and its uncertainty.

# Hypothesis Testing Workflow

A typical hypothesis testing process is:

### Step 1

Define the problem.

### Step 2

State the null hypothesis H₀.

### Step 3

State the alternative hypothesis Hₐ.

### Step 4

Choose the significance level α.

### Step 5

Select an appropriate statistical test.

### Step 6

Calculate the test statistic and p-value.

### Step 7

Compare the p-value with α.

### Step 8

Make a statistical decision.

### Step 9

Interpret the result in the business context.

# Business Example

A company launches a new website.

Before the change:

Average conversion rate = 5%

After the change:

Average conversion rate = 6%

The company wants to know whether the increase is statistically significant.

### Hypotheses

H₀: New conversion rate = 5%

Hₐ: New conversion rate > 5%

A suitable statistical test is selected based on the data and experimental design.

The result can help determine whether the observed increase is likely to reflect a real improvement rather than random sampling variation.

# Hypothesis Testing in AI/ML

Hypothesis testing is useful for:

- A/B testing
- Model comparison
- Feature analysis
- Experimentation
- Data analysis
- Business decision-making

### Example

Suppose Model A has accuracy:

90%

Model B has accuracy:

92%

The difference is 2 percentage points.

We should not automatically conclude that Model B is better.

Statistical analysis can help determine whether the observed difference is likely to be meaningful rather than simply due to sampling variation.

### Important

Statistical significance and practical/business significance are not always the same.

# One-Tailed vs Two-Tailed Tests

| Test | Purpose | Example |
|---|---|---|
| Left-tailed | Tests for a decrease | μ < 30 |
| Right-tailed | Tests for an increase | μ > 30 |
| Two-tailed | Tests for any difference | μ ≠ 30 |

## Key Idea

Use a one-tailed test only when the research question and alternative hypothesis are genuinely directional.

Use a two-tailed test when differences in either direction are relevant.

# Type I vs Type II Error

| Error | Meaning | Example |
|---|---|---|
| Type I Error | False Positive | Legitimate transaction flagged as fraud |
| Type II Error | False Negative | Fraudulent transaction not detected |

### Easy Way to Remember

Type I → We detect something that is not actually there.

Type II → We fail to detect something that actually exists.
