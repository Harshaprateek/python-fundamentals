```python
import numpy as np
from scipy import stats

group_a = np.array([72, 75, 78, 70, 74, 77, 73])
group_b = np.array([80, 82, 79, 85, 81, 83, 84])

t_statistic, p_value = stats.ttest_ind(
    group_a,
    group_b,
    equal_var=False
)

print("T-statistic:", t_statistic)
print("P-value:", p_value)

alpha = 0.05

if p_value < alpha:
    print("Reject the null hypothesis.")
else:
    print("Fail to reject the null hypothesis.")
```

### Interpretation

We compare the p-value with the significance level of 0.05.

If p-value < 0.05, there is statistically significant evidence of a difference between the group means.

If p-value ≥ 0.05, there is not sufficient evidence to conclude that the group means differ.

## 2. Z-Test

### Definition

A Z-Test is used to test hypotheses about population means or proportions when the assumptions for a z-test are appropriate.

For a one-sample mean z-test, the population standard deviation is typically known, or the sample is sufficiently large under the relevant assumptions.

### Formula

Z = (X̄ - μ) / (σ / √n)

Where:

- X̄ = Sample mean
- μ = Population mean under H₀
- σ = Population standard deviation
- n = Sample size

### Assumptions

- Observations are appropriately independent.
- Data is numerical for a mean z-test.
- Population standard deviation is known for the classical one-sample mean z-test.
- Normality or a suitable large-sample approximation is appropriate.

### When to Use

Use a z-test when its assumptions are satisfied.

### When Not to Use

Do not use it simply because the sample is large.

If the population standard deviation is unknown, a t-test is often more appropriate for a one-sample mean.

### Business Example

A manufacturer claims that the average product weight is 500 grams and the population standard deviation is known.

### AI/ML Example

Z-tests can be used in controlled experiments and statistical comparisons when their assumptions are satisfied.

```python
sample_mean = 510
population_mean = 500
population_std = 20
sample_size = 64

z_statistic = (
    sample_mean - population_mean
) / (
    population_std / np.sqrt(sample_size)
)

p_value = 2 * (1 - stats.norm.cdf(abs(z_statistic)))

print("Z-statistic:", z_statistic)
print("P-value:", p_value)

alpha = 0.05

if p_value < alpha:
    print("Reject the null hypothesis.")
else:
    print("Fail to reject the null hypothesis.")
```

### Interpretation

The z-statistic measures how many standard errors the sample mean is away from the hypothesized population mean.

The p-value is compared with the chosen significance level to make the statistical decision.

## 3. Chi-Square Test

### Definition

The Chi-Square Test is used with categorical data.

Two common applications are:

1. Test of Independence
2. Goodness-of-Fit Test

### Chi-Square Test of Independence

It determines whether two categorical variables are associated.

### Example

Is customer membership type related to purchase behavior?

Variables:

- Membership: Premium / Standard
- Purchase: Yes / No

### Assumptions

- Variables are categorical.
- Observations are appropriately independent.
- Expected frequencies should generally be sufficiently large for the standard approximation.

### When to Use

Use the test when examining relationships between categorical variables.

### When Not to Use

Do not use it directly for comparing means of numerical variables.

### Business Example

Determine whether membership type and product purchase are associated.

### AI/ML Example

Analyze relationships between categorical features during exploratory data analysis.

```python
observed = np.array([
    [30, 20],
    [15, 35]
])

chi2, p_value, degrees_of_freedom, expected = stats.chi2_contingency(
    observed
)

print("Chi-Square Statistic:", chi2)
print("P-value:", p_value)
print("Degrees of Freedom:", degrees_of_freedom)

print("\nExpected Frequencies:")
print(expected)

alpha = 0.05

if p_value < alpha:
    print("\nReject the null hypothesis.")
else:
    print("\nFail to reject the null hypothesis.")
```

### Interpretation

A small p-value provides evidence that the two categorical variables are associated.

A large p-value means there is not sufficient evidence of an association.

The test does not by itself establish causation.

## 4. ANOVA

### Definition

ANOVA stands for Analysis of Variance.

It is used to compare the means of three or more groups.

### Example

A company wants to compare average sales across:

- Region A
- Region B
- Region C

Instead of performing multiple t-tests, ANOVA can test whether there is evidence that at least one group mean differs.

### Hypotheses

H₀:

All group means are equal.

Hₐ:

At least one group mean is different.

### Assumptions

- Observations are appropriately independent.
- The outcome is numerical.
- The groups are reasonably suitable for ANOVA assumptions.
- Variances are reasonably similar for classical one-way ANOVA.

### When to Use

Use ANOVA when comparing three or more group means.

### When Not to Use

Do not use classical ANOVA when its assumptions are severely violated without considering an appropriate alternative.

### Business Example

Compare average customer spending across three marketing campaigns.

### AI/ML Example

Compare model performance across multiple experimental conditions.

```python
group_a = [70, 72, 75, 71, 73]
group_b = [78, 80, 82, 79, 81]
group_c = [85, 87, 84, 88, 86]

f_statistic, p_value = stats.f_oneway(
    group_a,
    group_b,
    group_c
)

print("F-statistic:", f_statistic)
print("P-value:", p_value)

alpha = 0.05

if p_value < alpha:
    print("Reject the null hypothesis.")
else:
    print("Fail to reject the null hypothesis.")
```

### Interpretation

If the p-value is less than 0.05, there is evidence that not all group means are equal.

ANOVA alone does not tell us exactly which groups differ.

Post-hoc tests can be used to investigate pairwise differences.

## 5. Mann-Whitney U Test

### Definition

The Mann-Whitney U Test is a non-parametric test used to compare two independent groups.

It is an alternative to the independent two-sample t-test when its assumptions are not suitable.

### Assumptions

- Two groups are independent.
- Observations are at least ordinal.
- Observations are appropriately independent within and between groups.

### When to Use

Use it when comparing two independent groups and a rank-based non-parametric method is appropriate.

### When Not to Use

Do not use it for:

- Paired observations
- More than two independent groups
- Dependent observations

### Business Example

Compare customer satisfaction scores between two stores.

### AI/ML Example

Compare model scores across two independent datasets when a non-parametric approach is appropriate.

```python
group_a = [70, 72, 75, 71, 73]
group_b = [80, 82, 79, 85, 81]

u_statistic, p_value = stats.mannwhitneyu(
    group_a,
    group_b,
    alternative="two-sided"
)

print("Mann-Whitney U Statistic:", u_statistic)
print("P-value:", p_value)

alpha = 0.05

if p_value < alpha:
    print("Reject the null hypothesis.")
else:
    print("Fail to reject the null hypothesis.")
```

### Interpretation

The test compares the distributions using ranks.

A small p-value provides evidence that the two groups differ in their distributions under the test's assumptions.

## 6. Shapiro-Wilk Test

### Definition

The Shapiro-Wilk Test is used to assess whether a sample is consistent with a normal distribution.

### Hypotheses

H₀:

The data is consistent with a normal distribution.

Hₐ:

The data is not consistent with a normal distribution.

### Assumptions

- Observations should be independent.
- The test is intended for assessing normality of numerical data.

### When to Use

Use it when you need a formal normality assessment and the sample size is appropriate for the test.

### When Not to Use

Do not rely only on the p-value.

For very large samples, tiny deviations from normality can become statistically significant.

Use visual methods such as:

- Histogram
- Q-Q plot

along with statistical tests.

### Business Example

Check whether measurement data is reasonably consistent with a normal distribution before using a statistical method that assumes normality.

### AI/ML Example

Check distributional assumptions during statistical analysis and model development.

```python
data = np.array([
    10.2, 11.1, 9.8, 10.5, 10.9,
    9.9, 10.3, 11.0, 10.1, 10.6
])

statistic, p_value = stats.shapiro(data)

print("Shapiro-Wilk Statistic:", statistic)
print("P-value:", p_value)

alpha = 0.05

if p_value < alpha:
    print("Reject the null hypothesis.")
    print("The data shows evidence of non-normality.")
else:
    print("Fail to reject the null hypothesis.")
    print("There is not sufficient evidence of non-normality.")
```

### Interpretation

A p-value below 0.05 provides evidence against the assumption that the data is normally distributed.

A p-value above 0.05 does not prove that the data is perfectly normal. It means there is not sufficient evidence to reject the normality assumption at the chosen significance level.

# Comparison of Statistical Tests

| Test | Main Purpose | Data Type |
|---|---|---|
| T-Test | Compare means | Numerical |
| Z-Test | Test mean/proportion under z-test assumptions | Numerical / Proportions |
| Chi-Square | Test categorical association | Categorical |
| ANOVA | Compare 3+ means | Numerical |
| Mann-Whitney U | Compare 2 independent groups using ranks | Ordinal / Numerical |
| Shapiro-Wilk | Assess consistency with normality | Numerical |

## Quick Selection Guide

### Comparing Two Means

→ T-Test

### Comparing Three or More Means

→ ANOVA

### Comparing Two Independent Groups with a Non-Parametric Approach

→ Mann-Whitney U

### Relationship Between Categorical Variables

→ Chi-Square

### Assessing Normality

→ Shapiro-Wilk

### Mean/Proportion Test Under Z-Test Assumptions

→ Z-Test

# How to Choose a Statistical Test

Before selecting a statistical test, ask:

### 1. What type of data do I have?

- Numerical
- Categorical
- Ordinal

### 2. How many groups are being compared?

- One
- Two
- Three or more

### 3. Are the groups independent or paired?

### 4. What assumptions are reasonable?

Check:

- Independence
- Distribution
- Variance
- Sample size

### 5. What is the business question?

The statistical test should be selected based on the research question and study design, not simply because a particular test is familiar.
