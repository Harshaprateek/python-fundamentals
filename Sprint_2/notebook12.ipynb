## 1. Correlation Matrix

### Definition

A correlation matrix shows the correlation between every pair of numerical features in a dataset.

Correlation values range from:

- +1 → Strong positive relationship
- 0 → No linear relationship
- -1 → Strong negative relationship

### Why is it important in ML?

A correlation matrix helps us:

- Understand relationships between features
- Detect redundant features
- Identify possible multicollinearity
- Perform Exploratory Data Analysis
- Support feature selection

### Example

Suppose we have:

- Age
- Experience
- Salary
- Spending Score

We can calculate the correlation between every pair of features.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

data = pd.DataFrame({
    "Age": [22, 25, 28, 32, 35, 40, 45, 50],
    "Experience": [1, 3, 5, 7, 9, 12, 15, 18],
    "Salary": [30000, 35000, 42000, 50000, 58000, 70000, 85000, 100000],
    "Spending_Score": [90, 85, 80, 75, 70, 60, 55, 50]
})

correlation_matrix = data.corr()

print(correlation_matrix)
```

### Interpretation

The correlation matrix shows how strongly the numerical features are related.

Features with correlation values close to +1 or -1 have strong linear relationships.

Strong relationships between predictor features may indicate possible multicollinearity.

## 2. Multicollinearity

### Definition

Multicollinearity occurs when two or more independent features in a machine learning model are highly correlated with each other.

### Example

Suppose a dataset contains:

- Years of Experience
- Years Since Graduation

These two features may contain very similar information.

### Why is it a Problem?

Multicollinearity can:

- Make regression coefficients unstable
- Make feature effects difficult to interpret
- Increase standard errors
- Make some statistical conclusions less reliable

### Important

Multicollinearity is especially important for models where coefficients are interpreted, such as linear and logistic regression.

Tree-based models are generally less sensitive to multicollinearity, although redundant features can still affect interpretation and efficiency.

### Business Example

Suppose a salary model uses:

- Years of Experience
- Years in Current Role

These variables may be strongly related.

### AI/ML Example

Before training a linear regression model, highly correlated predictor features may be investigated and potentially removed or transformed.

### Detecting Multicollinearity

Common approaches include:

1. Correlation Matrix
2. Variance Inflation Factor (VIF)

A correlation matrix provides a pairwise view.

VIF evaluates how much the variance of a regression coefficient is inflated because of relationships with other predictor variables.

## 3. Variance Inflation Factor (VIF)

### Definition

VIF stands for Variance Inflation Factor.

It measures how strongly a predictor variable is explained by the other predictor variables.

### Formula

VIF = 1 / (1 - R²)

Where:

R² is obtained by regressing one predictor against the other predictors.

### General Interpretation

- VIF = 1 → No multicollinearity
- VIF between 1 and 5 → Usually moderate
- VIF above 5 → Potential concern
- VIF above 10 → Often considered a strong warning

These are rules of thumb, not universal thresholds.

### Why Use VIF?

VIF is useful when analyzing regression models and investigating multicollinearity.

### Important

VIF should be interpreted together with the model, domain knowledge, sample size, and other diagnostics.

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor

X = data[[
    "Age",
    "Experience",
    "Salary",
    "Spending_Score"
]]

vif_data = pd.DataFrame()

vif_data["Feature"] = X.columns

vif_data["VIF"] = [
    variance_inflation_factor(X.values, i)
    for i in range(X.shape[1])
]

print(vif_data)
```

### Interpretation

VIF values help identify features that may have strong relationships with the other predictor variables.

A high VIF indicates that a feature can be strongly explained by the remaining predictors.

VIF should be used as a diagnostic rather than an automatic reason to remove a feature.

## 4. Feature Importance

### Definition

Feature importance measures how useful individual features are for making predictions in a particular model.

Different machine learning models calculate or estimate feature importance differently.

### Example

Suppose we predict customer churn using:

- Age
- Monthly Charges
- Contract Type
- Tenure

The model may determine that Monthly Charges and Tenure are particularly useful for prediction.

### Why is Feature Importance Useful?

It can help with:

- Understanding models
- Feature analysis
- Feature selection
- Model interpretation
- Business decision-making

### Important

Feature importance does not automatically mean that a feature causes the prediction.

It describes the feature's contribution or usefulness according to the specific model and importance method.

### Interpretation

The feature importance values show how much each feature contributes to the Random Forest model's predictive decisions according to its built-in importance measure.

Higher importance means the feature was more useful to the model under this particular calculation.

Feature importance can change when:

- The dataset changes
- The model changes
- Hyperparameters change
- Different importance methods are used

## 5. Feature Selection Basics

### Definition

Feature Selection is the process of selecting the most useful features for a machine learning model.

The goal is not simply to keep as many features as possible.

### Why Feature Selection?

Feature selection can help:

- Reduce unnecessary features
- Reduce computational cost
- Improve model interpretability
- Reduce noise
- Reduce overfitting in some situations
- Improve model training efficiency

### Types of Feature Selection

#### 1. Filter Methods

Select features using statistical properties.

Examples:

- Correlation
- Chi-Square
- Mutual Information

#### 2. Wrapper Methods

Evaluate different feature subsets using a model.

Examples:

- Recursive Feature Elimination (RFE)

#### 3. Embedded Methods

Feature selection happens as part of model training.

Examples:

- Lasso Regression
- Tree-based feature importance

### Correlation-Based Selection

Suppose two predictor features have a very high correlation.

Example:

Feature A ↔ Feature B

Correlation = 0.95

They may contain highly overlapping information.

We can investigate whether both features are necessary.

### Important

Do not automatically remove a feature only because correlation is high.

Consider:

- Business meaning
- Model type
- Target relationship
- Multicollinearity
- Model performance

```python
feature_correlations = data.corr()

print("Correlation with Salary:")

print(
    feature_correlations["Salary"]
    .sort_values(ascending=False)
)
```

## Recursive Feature Elimination (RFE)

RFE is a wrapper-based feature selection method.

### How It Works

1. Train a model using all features.
2. Determine feature importance.
3. Remove the least important feature.
4. Train the model again.
5. Repeat until the desired number of features remains.

### Advantage

It considers the relationship between features and the selected model.

### Limitation

It can be computationally expensive when there are many features.

# Feature Selection Workflow

A practical feature selection process can be:

### Step 1 — Understand the Data

Understand what each feature represents.

### Step 2 — Remove Clearly Invalid Features

Remove features that are:

- Duplicates
- Incorrect
- Constant
- Irrelevant

### Step 3 — Analyze Relationships

Use:

- Correlation
- VIF
- Domain knowledge

### Step 4 — Train a Baseline Model

Measure model performance before feature selection.

### Step 5 — Apply Feature Selection

Use an appropriate:

- Filter method
- Wrapper method
- Embedded method

### Step 6 — Retrain

Train the model using the selected features.

### Step 7 — Compare Performance

Compare the original and reduced feature sets using appropriate validation.

### Step 8 — Choose the Practical Solution

Consider:

- Performance
- Interpretability
- Computational cost
- Business requirements
