# 1. Histogram

## Purpose

A histogram shows the distribution of numerical data by dividing values into ranges called bins.

## When to Use

- Understand data distribution
- Detect skewness
- Identify unusual values
- Understand frequency

## Advantages

- Easy to understand
- Shows distribution clearly
- Helps identify skewness

## Limitations

- Sensitive to bin size
- Not suitable for categorical data

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)

data = np.random.normal(50, 10, 500)

plt.hist(data, bins=20)
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram")
plt.show()
```

# 2. Bar Chart

## Purpose

A bar chart compares values across different categories.

## When to Use

- Compare categories
- Compare sales by product
- Compare customers by region

## Advantages

- Simple
- Easy to compare categories
- Works well with categorical data

## Limitations

- Too many categories make it difficult to read
- Not ideal for continuous numerical data

```python
categories = ["A", "B", "C", "D"]
values = [25, 40, 30, 50]

plt.bar(categories, values)
plt.xlabel("Category")
plt.ylabel("Value")
plt.title("Bar Chart")
plt.show()
```

# 3. Line Chart

## Purpose

A line chart shows trends or changes over time.

## When to Use

- Time-series data
- Monthly sales
- Stock prices
- Website traffic

## Advantages

- Clearly shows trends
- Good for time-series data

## Limitations

- Not ideal for categorical comparison
- Too many lines can become confusing

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [100, 120, 115, 150, 170, 190]

plt.plot(months, sales, marker="o")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Monthly Sales")
plt.show()
```

# 4. Scatter Plot

## Purpose

A scatter plot shows the relationship between two numerical variables.

## When to Use

- Find correlation
- Identify patterns
- Detect outliers

## Advantages

- Shows relationships clearly
- Helps identify correlation
- Helps detect outliers

## Limitations

- Difficult to interpret with very large datasets
- Only shows relationships between variables

```python
hours = [1, 2, 3, 4, 5, 6, 7, 8]
scores = [45, 50, 55, 60, 65, 72, 78, 85]

plt.scatter(hours, scores)
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.title("Study Hours vs Exam Score")
plt.show()
```

# 5. Pie Chart

## Purpose

A pie chart shows how a total is divided among categories.

## When to Use

- Market share
- Budget distribution
- Percentage breakdown

## Advantages

- Easy to understand for a small number of categories
- Shows proportions

## Limitations

- Difficult to compare similar-sized categories
- Not suitable for many categories

```python
labels = ["Python", "Java", "C++", "JavaScript"]
sizes = [40, 25, 20, 15]

plt.pie(sizes, labels=labels, autopct="%1.1f%%")
plt.title("Programming Language Usage")
plt.show()
```

# 6. Density Plot

## Purpose

A density plot shows the smooth distribution of numerical data.

## When to Use

- Understand probability distributions
- Compare distributions
- Identify peaks and skewness

## Advantages

- Smooth representation
- Useful for comparing distributions

## Limitations

- Can hide individual observations
- Bandwidth selection affects the shape

```python
data = np.random.normal(50, 10, 500)

sns.kdeplot(data=data, fill=True)
plt.xlabel("Value")
plt.title("Density Plot")
plt.show()
```

# 7. Box Plot

## Purpose

A box plot summarizes the distribution of numerical data using:

- Minimum
- Q1
- Median
- Q3
- Maximum
- Outliers

## When to Use

- Detect outliers
- Compare distributions
- Understand spread

## Advantages

- Quickly identifies outliers
- Shows median and spread

## Limitations

- Does not show the exact distribution shape

```python
data = [10, 12, 14, 15, 16, 18, 20, 22, 25, 50]

plt.boxplot(data)
plt.ylabel("Value")
plt.title("Box Plot")
plt.show()
```

# 8. Violin Plot

## Purpose

A violin plot combines a box plot with a density plot.

## When to Use

- Compare distributions
- Understand data spread
- Analyze multiple groups

## Advantages

- Shows distribution shape
- Shows central tendency and spread

## Limitations

- More difficult to interpret than a box plot
- Requires enough data for meaningful density estimation

```python
df = pd.DataFrame({
    "Department": ["IT"] * 50 + ["HR"] * 50,
    "Salary": list(np.random.normal(70000, 8000, 50)) +
              list(np.random.normal(60000, 6000, 50))
})

sns.violinplot(data=df, x="Department", y="Salary")
plt.title("Salary Distribution by Department")
plt.show()
```

# 9. Pair Plot

## Purpose

A pair plot visualizes relationships between multiple numerical variables.

## When to Use

- Exploratory Data Analysis (EDA)
- Finding correlations
- Comparing multiple features

## Advantages

- Provides many relationships in one visualization
- Useful for EDA

## Limitations

- Can become difficult to read with many features
- Computationally expensive for large datasets

```python
df = sns.load_dataset("iris")

sns.pairplot(df, hue="species")
plt.show()
```

# 10. Heatmap

## Purpose

A heatmap represents numerical values using different levels of intensity.

It is commonly used to visualize correlation matrices.

## When to Use

- Correlation analysis
- Feature relationships
- Finding multicollinearity

## Advantages

- Easy to identify strong relationships
- Compact representation

## Limitations

- Large matrices can become difficult to read
- Correlation does not imply causation

```python
df = pd.DataFrame({
    "Age": [20, 25, 30, 35, 40],
    "Salary": [30000, 40000, 50000, 60000, 70000],
    "Experience": [1, 2, 4, 6, 8],
    "Score": [60, 65, 72, 80, 88]
})

correlation = df.corr()

sns.heatmap(correlation, annot=True)
plt.title("Correlation Heatmap")
plt.show()
```
