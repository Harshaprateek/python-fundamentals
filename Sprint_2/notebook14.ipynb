# 1. Statistics in Data Cleaning

Statistics helps us understand the structure and quality of a dataset.

Common statistical techniques include:

- Mean
- Median
- Mode
- Standard deviation
- Variance
- Quartiles
- Percentiles
- Correlation

These techniques help identify unusual or incorrect data.

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Age": [20, 22, 25, 28, 30, 35, 40],
    "Salary": [30000, 35000, 40000, 45000, 50000, 60000, 70000]
})

print(df.describe())
```

## Interpretation

The `describe()` function provides statistical information such as:

- Count
- Mean
- Standard deviation
- Minimum
- Quartiles
- Maximum

This is useful during Exploratory Data Analysis.

# 2. Missing Value Handling

Statistics can help us decide how to replace missing values.

Common methods:

- Mean
- Median
- Mode

### Mean

Useful when the data is approximately symmetric.

### Median

Useful when the data contains outliers or is skewed.

### Mode

Useful for categorical variables.

```python
df = pd.DataFrame({
    "Age": [20, 25, np.nan, 30, 35],
    "Salary": [30000, 40000, 50000, np.nan, 60000]
})

df["Age"] = df["Age"].fillna(df["Age"].median())
df["Salary"] = df["Salary"].fillna(df["Salary"].median())

print(df)
```

# 3. Outlier Detection

Statistics helps identify observations that are very different from the rest of the data.

Common methods:

- IQR
- Z-score
- Box Plot

### IQR

IQR = Q3 - Q1

An observation can be considered an outlier if it is:

Below Q1 - 1.5 × IQR

or

Above Q3 + 1.5 × IQR

```python
data = np.array([10, 12, 14, 15, 16, 18, 20, 50])

Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = data[(data < lower) | (data > upper)]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Outliers:", outliers)
```

# 4. Feature Engineering

Statistics helps us create useful features from existing data.

Examples:

- Mean
- Sum
- Ratio
- Percentage
- Log transformation
- Aggregation
- Interaction features

Example:

Instead of using only total sales, we can create:

Average Monthly Sales = Total Sales / Number of Months

```python
df = pd.DataFrame({
    "Total_Sales": [10000, 20000, 30000],
    "Months": [10, 10, 10]
})

df["Average_Monthly_Sales"] = df["Total_Sales"] / df["Months"]

print(df)
```

# 5. Feature Scaling

Feature scaling makes numerical features comparable.

Two common techniques:

1. Standardization
2. Min-Max Scaling

### Standardization

Formula:

z = (x - mean) / standard deviation

After standardization, the feature generally has:

Mean ≈ 0

Standard deviation ≈ 1

```python
from sklearn.preprocessing import StandardScaler

data = [[10], [20], [30], [40], [50]]

scaler = StandardScaler()

scaled_data = scaler.fit_transform(data)

print(scaled_data)
```

# 6. Data Normalization

Normalization transforms values into a specific range, commonly 0 to 1.

### Formula

x' = (x - min) / (max - min)

Normalization is useful when features have different scales.

```python
from sklearn.preprocessing import MinMaxScaler

data = [[10], [20], [30], [40], [50]]

scaler = MinMaxScaler()

normalized_data = scaler.fit_transform(data)

print(normalized_data)
```

# 7. Statistics in Model Evaluation

Statistics helps evaluate Machine Learning models.

### Regression Metrics

- MAE
- MSE
- RMSE
- R²

### Classification Metrics

- Accuracy
- Precision
- Recall
- F1-score
- ROC-AUC

Statistical evaluation helps determine whether a model performs well on unseen data.

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

actual = [100, 200, 300, 400]
predicted = [110, 190, 310, 380]

mae = mean_absolute_error(actual, predicted)
mse = mean_squared_error(actual, predicted)
rmse = np.sqrt(mse)
r2 = r2_score(actual, predicted)

print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
print("R2:", r2)
```

# 8. Statistics in Linear Regression

Linear Regression models the relationship between dependent and independent variables.

General equation:

y = β0 + β1x + ε

Where:

- y = dependent variable
- x = independent variable
- β0 = intercept
- β1 = coefficient
- ε = error

Statistics is used to understand:

- Coefficients
- Residuals
- R²
- Error
- Relationships between variables

```python
from sklearn.linear_model import LinearRegression

X = np.array([[1], [2], [3], [4], [5]])
y = np.array([10, 20, 30, 40, 50])

model = LinearRegression()
model.fit(X, y)

print("Coefficient:", model.coef_[0])
print("Intercept:", model.intercept_)
print("R2 Score:", model.score(X, y))
```

# 9. Statistics in Logistic Regression

Logistic Regression is commonly used for classification.

It predicts probabilities between 0 and 1.

Example:

Predict whether a customer will:

- Buy a product
- Not buy a product

Statistics helps understand:

- Probability
- Log-odds
- Coefficients
- Classification performance

```python
from sklearn.linear_model import LogisticRegression

X = np.array([[1], [2], [3], [4], [5], [6]])
y = np.array([0, 0, 0, 1, 1, 1])

model = LogisticRegression()
model.fit(X, y)

prediction = model.predict([[4]])
probability = model.predict_proba([[4]])

print("Prediction:", prediction)
print("Probability:", probability)
```

# 10. Statistics in Naive Bayes

Naive Bayes is based on Bayes' theorem.

Bayes' theorem:

P(A|B) = P(B|A) × P(A) / P(B)

It is commonly used for:

- Text classification
- Spam detection
- Sentiment analysis
- Document classification

The algorithm calculates probabilities for different classes and selects the most probable class.

```python
from sklearn.naive_bayes import GaussianNB

X = np.array([
    [1, 20],
    [2, 25],
    [3, 30],
    [8, 70],
    [9, 75],
    [10, 80]
])

y = np.array([0, 0, 0, 1, 1, 1])

model = GaussianNB()
model.fit(X, y)

prediction = model.predict([[4, 35]])

print("Prediction:", prediction)
```

# 11. Statistics in Clustering

Clustering groups similar observations together.

Example:

Customer segmentation.

Statistical concepts used include:

- Distance
- Mean
- Variance
- Distribution
- Similarity

K-Means clustering calculates the mean position of each cluster, called the centroid.

```python
from sklearn.cluster import KMeans

X = np.array([
    [1, 2],
    [1, 3],
    [2, 2],
    [8, 8],
    [9, 8],
    [8, 9]
])

model = KMeans(n_clusters=2, random_state=42, n_init=10)

model.fit(X)

print("Cluster Labels:", model.labels_)
print("Centroids:")
print(model.cluster_centers_)
```

# 12. Statistics in PCA

PCA stands for Principal Component Analysis.

PCA reduces the number of features while preserving as much useful information as possible.

Statistics involved:

- Mean
- Variance
- Covariance
- Eigenvectors
- Eigenvalues

PCA is commonly used for:

- Dimensionality reduction
- Visualization
- Noise reduction
- Feature compression

```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

X = np.array([
    [10, 100],
    [20, 200],
    [30, 300],
    [40, 400],
    [50, 500]
])

X_scaled = StandardScaler().fit_transform(X)

pca = PCA(n_components=1)

X_pca = pca.fit_transform(X_scaled)

print("Original shape:", X.shape)
print("Reduced shape:", X_pca.shape)
print("Explained variance:", pca.explained_variance_ratio_)
```
