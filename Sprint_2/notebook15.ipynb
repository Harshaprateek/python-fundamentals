## 1. Explain the difference between Mean and Median

**Mean** is the average of all values.

**Formula:**

Mean = Sum of all values / Number of values

**Median** is the middle value after arranging the data in ascending order.

**Example:**

Data: 10, 20, 30, 40, 100

- Mean = 40
- Median = 30

**Key Difference:** Mean is affected by outliers, while Median is less affected by outliers.

---

## 2. What is Variance?

**Variance** measures how much the data values are spread out from the mean.

- Low variance → Values are close to the mean.
- High variance → Values are more spread out.

Variance is calculated by taking the average of the squared differences between each value and the mean.

---

## 3. What is Standard Deviation?

**Standard Deviation** measures the spread of data around the mean.

It is the square root of variance.

- Low standard deviation → Data is close to the mean.
- High standard deviation → Data is widely spread.

**Example:**

If students' marks have a low standard deviation, most students scored close to the average.

---

## 4. Explain Bayes' Theorem with an example

**Bayes' Theorem** calculates the probability of an event based on previously known information.

**Formula:**

P(A|B) = P(B|A) × P(A) / P(B)

**Example:**

Suppose a medical test is used to detect a disease.

If a person tests positive, Bayes' Theorem can help calculate the probability that the person actually has the disease.

**AI/ML Example:**

Naive Bayes uses Bayes' Theorem for tasks such as:

- Spam detection
- Text classification
- Sentiment analysis

---

## 5. Difference between Population and Sample

**Population** means the entire group that we want to study.

**Sample** is a smaller subset selected from the population.

**Example:**

- All students in a university → Population
- 500 selected students → Sample

We use samples because studying the entire population can be expensive and time-consuming.

---

## 6. Explain Correlation and Covariance

**Covariance** tells us the direction of the relationship between two variables.

- Positive covariance → Variables tend to increase together.
- Negative covariance → One variable tends to increase while the other decreases.

**Correlation** measures both the direction and strength of the relationship.

Correlation ranges from **-1 to +1**.

- +1 → Strong positive relationship
- -1 → Strong negative relationship
- 0 → No linear relationship

**Key Difference:** Covariance shows the direction of a relationship but is scale-dependent. Correlation is standardized between -1 and +1.

---

## 7. What is Skewness?

**Skewness** measures the asymmetry of a data distribution.

There are three common cases:

- Skewness ≈ 0 → Approximately symmetric
- Positive skewness → Longer tail on the right
- Negative skewness → Longer tail on the left

**Example:**

Income data often has positive skewness because a small number of people may have extremely high incomes.

---

## 8. Explain IQR

**IQR** stands for **Interquartile Range**.

It measures the spread of the middle 50% of the data.

**Formula:**

IQR = Q3 - Q1

Where:

- Q1 = 25th percentile
- Q3 = 75th percentile

IQR is commonly used to detect outliers.

**Outlier Boundaries:**

Lower Bound = Q1 - 1.5 × IQR

Upper Bound = Q3 + 1.5 × IQR

Values below the lower bound or above the upper bound can be considered outliers.

---

## 9. What is a p-value?

A **p-value** helps determine whether the observed result provides enough evidence against the null hypothesis.

We usually compare the p-value with a significance level such as **0.05**.

- p-value < 0.05 → Reject the null hypothesis
- p-value ≥ 0.05 → Fail to reject the null hypothesis

**Important:** A p-value is not the probability that the null hypothesis is true.

---

## 10. Difference between Type I and Type II Error

**Type I Error** occurs when we reject a true null hypothesis.

It is commonly called a **False Positive**.

**Type II Error** occurs when we fail to reject a false null hypothesis.

It is commonly called a **False Negative**.

| Error | Meaning |
|---|---|
| Type I Error | False Positive |
| Type II Error | False Negative |

**Example:**

In a medical test:

- Type I Error → Say a healthy person has the disease.
- Type II Error → Say a sick person does not have the disease.

````python
import numpy as np

data = np.array([10, 12, 14, 15, 16, 18, 20, 22, 25, 50])

Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = data[
    (data < lower_bound) | (data > upper_bound)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Outliers:", outliers)
````
