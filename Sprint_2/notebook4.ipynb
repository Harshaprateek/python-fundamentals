1. Normal Distribution

## 1. Normal Distribution

### Understand the Concept

A Normal Distribution is a continuous probability distribution where most values are concentrated around the mean.

It has a bell-shaped curve.

Important characteristics:

- Mean = Median = Mode
- The distribution is symmetric
- Most observations are near the center
- Fewer observations occur farther from the mean

The normal distribution is very common in statistics and machine learning.

### Example

Suppose the heights of students in a large population are approximately normally distributed.

Most students will have heights close to the average, while very short and very tall students will be less common.

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

data = np.random.normal(loc=170, scale=10, size=1000)

plt.hist(data, bins=30, density=True, edgecolor="black")
plt.title("Normal Distribution")
plt.xlabel("Height")
plt.ylabel("Density")
plt.show()
```

2. Uniform Distribution

## 2. Uniform Distribution

### Understand the Concept

In a Uniform Distribution, every value within a specified range has an equal probability of occurring.

There are two types:

- Discrete Uniform Distribution
- Continuous Uniform Distribution

For a continuous uniform distribution, values between a minimum and maximum are equally likely.

### Example

Rolling a fair six-sided die is an example of a discrete uniform distribution because every number from 1 to 6 has the same probability.

```python
data = np.random.uniform(0, 10, 1000)

plt.hist(data, bins=20, edgecolor="black")
plt.title("Uniform Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

### Interpretation

The values are spread relatively evenly between 0 and 10.

Unlike the normal distribution, there is no central value where observations are strongly concentrated.

## 3. Gaussian Distribution

### Understand the Concept

Gaussian Distribution is another name for the Normal Distribution.

It is defined using two important parameters:

- Mean (μ)
- Standard Deviation (σ)

The probability density function is:

f(x) = (1 / (σ√(2π))) × e^(-(x-μ)² / (2σ²))

Where:

- μ = mean
- σ = standard deviation
- x = observation

### AI/ML Example

Many machine learning algorithms use assumptions related to Gaussian distributions.

Gaussian noise is also commonly used when testing machine learning models.

```python
mean = 0
std = 1

data = np.random.normal(mean, std, 1000)

plt.hist(data, bins=30, density=True, edgecolor="black")
plt.title("Gaussian Distribution")
plt.xlabel("Value")
plt.ylabel("Density")
plt.show()
```

## 4. Left Skew (Negative Skew)

### Understand the Concept

A left-skewed distribution has a longer tail toward the left side.

It is also called a negatively skewed distribution.## 4. Left Skew (Negative Skew)

### Understand the Concept

A left-skewed distribution has a longer tail toward the left side.

It is also called a negatively skewed distribution.

Usually:

Mean < Median < Mode

### Example

Suppose most students score very high on an easy exam, but a few students receive very low scores.

The low scores create a tail toward the left.


```python
data = 100 - np.random.exponential(scale=10, size=1000)

plt.hist(data, bins=30, edgecolor="black")
plt.title("Left-Skewed Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

## 5. Right Skew (Positive Skew)

### Understand the Concept

A right-skewed distribution has a longer tail toward the right side.

It is also called a positively skewed distribution.

Usually:

Mode < Median < Mean

### Example

Income is often right-skewed.

Most people may have moderate incomes, while a small number of people have extremely high incomes.

Those high values create a long right tail.

```python
data = np.random.exponential(scale=2, size=1000)

plt.hist(data, bins=30, edgecolor="black")
plt.title("Right-Skewed Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

## 6. Symmetric Distribution

### Understand the Concept

A distribution is symmetric when the left side and right side have approximately the same shape.

In a perfectly symmetric distribution:

Mean = Median = Mode

The Normal Distribution is a common example of a symmetric distribution.

```python
data = np.random.normal(50, 10, 1000)

plt.hist(data, bins=30, edgecolor="black")
plt.axvline(np.mean(data), linestyle="--", label="Mean")
plt.axvline(np.median(data), linestyle=":", label="Median")

plt.title("Symmetric Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```

## 7. Kurtosis

### Understand the Concept

Kurtosis describes the shape of the tails of a distribution and how much extreme values are present.

There are three common categories:

### 1. Mesokurtic

Similar to a normal distribution.

### 2. Leptokurtic

Has heavier tails and potentially more extreme values.

### 3. Platykurtic

Has lighter tails and fewer extreme values.

Kurtosis is useful for identifying distributions that may contain extreme observations.

```python
from scipy.stats import kurtosis

data = np.random.normal(0, 1, 1000)

kurtosis_value = kurtosis(data)

print("Kurtosis:", kurtosis_value)
```

## 8. Empirical Rule (68–95–99.7 Rule)

### Understand the Concept

The Empirical Rule applies to approximately normally distributed data.

It states:

- Approximately 68% of data lies within ±1 standard deviation of the mean.
- Approximately 95% lies within ±2 standard deviations.
- Approximately 99.7% lies within ±3 standard deviations.

For example, if:

Mean = 100

Standard Deviation = 10

Then:

### Within 1 standard deviation

90 to 110 → approximately 68%

### Within 2 standard deviations

80 to 120 → approximately 95%

### Within 3 standard deviations

70 to 130 → approximately 99.7%

```python
mean = 100
std = 10

print("Mean:", mean)
print("Standard Deviation:", std)

print("\nWithin 1 Standard Deviation:")
print(mean - std, "to", mean + std)

print("\nWithin 2 Standard Deviations:")
print(mean - 2 * std, "to", mean + 2 * std)

print("\nWithin 3 Standard Deviations:")
print(mean - 3 * std, "to", mean + 3 * std)
```
