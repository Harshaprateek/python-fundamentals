## 1. Bernoulli Distribution

### Definition

Bernoulli Distribution is used for an experiment that has exactly two possible outcomes.

The two outcomes are usually represented as:

- 1 → Success
- 0 → Failure

### Formula

P(X = x) = p^x × (1-p)^(1-x)

Where:

- p = Probability of success
- x = 0 or 1
- 1-p = Probability of failure

### Example

Suppose a customer either:

- Purchases a product → Success
- Does not purchase → Failure

If the probability of purchase is 0.7:

P(Success) = 0.7

P(Failure) = 0.3

### Where Used

- Yes/No outcomes
- Success/Failure experiments
- Click/No-click prediction
- Fraud/Not fraud
- Customer purchase prediction

### AI/ML Example

Binary classification problems can be modeled using probabilities similar to Bernoulli outcomes.

Examples:

- Spam / Not Spam
- Fraud / Not Fraud
- Churn / Not Churn

```python
import numpy as np
import matplotlib.pyplot as plt

p = 0.7

outcomes = [0, 1]
probabilities = [1 - p, p]

plt.bar(outcomes, probabilities)
plt.xlabel("Outcome")
plt.ylabel("Probability")
plt.title("Bernoulli Distribution")
plt.xticks([0, 1], ["Failure", "Success"])
plt.show()
```

## 2. Binomial Distribution

### Definition

Binomial Distribution represents the number of successes in a fixed number of independent trials.

Each trial has only two possible outcomes:

- Success
- Failure

### Formula

P(X = k) = C(n,k) × p^k × (1-p)^(n-k)

Where:

- n = Number of trials
- k = Number of successes
- p = Probability of success
- C(n,k) = Combination

### Example

Suppose a salesperson contacts 10 customers.

The probability that a customer purchases a product is 0.3.

We can use the Binomial Distribution to calculate the probability of getting exactly 3 purchases.

### Where Used

- Number of successful sales
- Number of defective products
- Number of successful predictions
- Survey responses
- Conversion analysis

### AI/ML Example

It can be used when counting binary outcomes across a fixed number of observations.

```python
from scipy.stats import binom

n = 10
p = 0.3

x = np.arange(0, n + 1)
probabilities = binom.pmf(x, n, p)

plt.bar(x, probabilities)
plt.xlabel("Number of Successes")
plt.ylabel("Probability")
plt.title("Binomial Distribution")
plt.show()
```

## 3. Poisson Distribution

### Definition

Poisson Distribution is used to model the number of times an event occurs within a fixed interval of time or space.

### Formula

P(X = k) = (λ^k × e^(-λ)) / k!

Where:

- λ = Average number of events
- k = Number of events
- e = Euler's number

### Example

Suppose a customer support center receives an average of 5 calls per hour.

Poisson Distribution can estimate the probability of receiving exactly 7 calls in one hour.

### Where Used

- Number of customer calls
- Website visits
- System failures
- Accidents
- Arrivals at a service center

### AI/ML Example

Poisson Distribution can be useful for modeling count-based events such as:

- Number of incidents
- Number of requests
- Number of events occurring within a time period

```python
from scipy.stats import poisson

lambda_value = 5

x = np.arange(0, 15)
probabilities = poisson.pmf(x, lambda_value)

plt.bar(x, probabilities)
plt.xlabel("Number of Events")
plt.ylabel("Probability")
plt.title("Poisson Distribution")
plt.show()
```

## 4. Uniform Distribution

### Definition

In a Uniform Distribution, all values within a specified range have equal probability.

There are two types:

- Discrete Uniform Distribution
- Continuous Uniform Distribution

### Formula

For a continuous uniform distribution:

f(x) = 1 / (b-a)

Where:

- a = Lower limit
- b = Upper limit

### Example

Suppose a random number is selected between 0 and 10.

Every value in that range has equal likelihood.

### Where Used

- Random number generation
- Simulation
- Random sampling
- Testing

### AI/ML Example

Uniform distributions are often used for:

- Random initialization
- Simulation
- Data generation
- Hyperparameter initialization

```python
from scipy.stats import uniform

x = np.linspace(0, 10, 100)
y = uniform.pdf(x, loc=0, scale=10)

plt.plot(x, y)
plt.xlabel("Value")
plt.ylabel("Density")
plt.title("Uniform Distribution")
plt.show()
```

## 5. Normal Distribution

### Definition

Normal Distribution is one of the most important probability distributions in statistics.

It is symmetric around the mean and has a bell-shaped curve.

### Formula

f(x) = (1 / (σ√(2π))) × e^(-(x-μ)² / (2σ²))

Where:

- μ = Mean
- σ = Standard deviation

### Properties

- Mean = Median = Mode
- Symmetric around the mean
- Bell-shaped curve
- Most observations are close to the mean

### Example

Human heights can approximately follow a normal distribution.

### Where Used

- Measurement data
- Quality control
- Statistical analysis
- Error modeling
- Forecasting

### AI/ML Example

Normal Distribution is important in:

- Feature analysis
- Statistical assumptions
- Gaussian models
- Linear regression
- Gaussian Naive Bayes
- Data preprocessing

```python
from scipy.stats import norm

mean = 0
std = 1

x = np.linspace(-4, 4, 100)

y = norm.pdf(x, mean, std)

plt.plot(x, y)
plt.xlabel("Value")
plt.ylabel("Density")
plt.title("Normal Distribution")
plt.show()
```

## 6. Exponential Distribution

### Definition

Exponential Distribution is commonly used to model the time between events.

It is related to the Poisson Distribution.

### Formula

f(x) = λe^(-λx)

Where:

- λ = Rate parameter
- x = Time between events

### Example

Suppose a customer support center receives calls at a constant rate.

The Exponential Distribution can model the waiting time until the next call.

### Where Used

- Waiting time
- Time between arrivals
- Machine failure time
- Customer service
- Network requests

### AI/ML Example

It can be used in:

- Survival analysis
- Reliability modeling
- Event-time prediction
- Queueing models

### Important Difference

Poisson Distribution:

Counts **how many events** occur.

Exponential Distribution:

Measures **how much time between events**.

```python
from scipy.stats import expon

scale = 2

x = np.linspace(0, 10, 100)
y = expon.pdf(x, scale=scale)

plt.plot(x, y)
plt.xlabel("Time")
plt.ylabel("Density")
plt.title("Exponential Distribution")
plt.show()
```
