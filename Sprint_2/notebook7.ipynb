## 1. Sampling

### Definition

Sampling is the process of selecting a smaller group of observations from a larger population.

### Population

A population is the complete group that we want to study.

Example:

All customers of a company.

### Sample

A sample is a smaller group selected from the population.

Example:

1,000 customers selected from 100,000 customers.

### Why Sampling is Important

Studying the entire population can be:

- Expensive
- Time-consuming
- Difficult to manage

Sampling allows us to analyze a smaller amount of data while trying to represent the larger population.

### Business Example

A company may survey 1,000 customers instead of asking all 1 million customers.

### AI/ML Example

Instead of training or testing on an extremely large dataset during experimentation, data scientists may use representative samples for faster analysis.

```python
import numpy as np
import pandas as pd

population = np.arange(1, 101)

sample = np.random.choice(population, size=10, replace=False)

print("Population:", population)
print("Sample:", sample)
```

## 2. Random Sampling

### Definition

Random Sampling is a method where every member of the population has an equal chance of being selected.

### Example

Suppose a company has 10,000 customers.

We randomly select 500 customers for a survey.

Each customer has an equal chance of being selected.

### Advantages

- Simple
- Reduces selection bias
- Easy to implement

### Limitations

- May not represent small subgroups properly
- Requires access to the population

### Business Example

Randomly selecting customers to measure customer satisfaction.

### AI/ML Example

Randomly splitting data into training and testing datasets.

```python
population = np.arange(1, 101)

random_sample = np.random.choice(
    population,
    size=10,
    replace=False
)

print("Random Sample:")
print(random_sample)
```

## 3. Stratified Sampling

### Definition

Stratified Sampling divides the population into smaller groups called strata and then selects samples from each group.

The groups are usually created based on an important characteristic.

### Example

A company's customers can be divided into:

- Premium
- Standard
- Basic

We then select customers from each category.

### Why Use It?

Stratified Sampling ensures that important subgroups are represented in the sample.

### Business Example

A company wants customer feedback from different age groups.

It can create groups:

- 18–30
- 31–45
- 46–60
- 60+

Then sample customers from each group.

### AI/ML Example

Stratified sampling is commonly used when splitting classification datasets so that different classes maintain similar proportions.

```python
data = pd.DataFrame({
    "Customer_ID": range(1, 11),
    "Category": [
        "Premium", "Premium",
        "Standard", "Standard", "Standard",
        "Basic", "Basic", "Basic", "Basic", "Basic"
    ]
})

print(data)

stratified_sample = (
    data.groupby("Category", group_keys=False)
    .apply(lambda x: x.sample(min(2, len(x))))
)

print("\nStratified Sample:")
print(stratified_sample)
```

## 4. Cluster Sampling

### Definition

Cluster Sampling divides the population into natural groups called clusters.

Instead of selecting individuals from every cluster, we randomly select some clusters and study the members within those clusters.

### Example

A university has 20 departments.

Instead of selecting students from every department, we randomly select 5 departments and survey students from those departments.

### Business Example

A company has stores across many cities.

Instead of surveying customers from every store, it selects some stores and surveys customers from those stores.

### Advantages

- Cost-effective
- Useful for geographically distributed populations
- Easier to organize

### Limitation

Selected clusters may not perfectly represent the entire population.

```python
data = pd.DataFrame({
    "Customer_ID": range(1, 13),
    "City": [
        "Hyderabad", "Hyderabad", "Hyderabad",
        "Mumbai", "Mumbai", "Mumbai",
        "Delhi", "Delhi", "Delhi",
        "Chennai", "Chennai", "Chennai"
    ]
})

selected_clusters = np.random.choice(
    data["City"].unique(),
    size=2,
    replace=False
)

cluster_sample = data[
    data["City"].isin(selected_clusters)
]

print("Selected Clusters:")
print(selected_clusters)

print("\nCluster Sample:")
print(cluster_sample)
```

## 5. Systematic Sampling

### Definition

Systematic Sampling selects observations at regular intervals from an ordered population.

### Formula

Sampling Interval:

k = Population Size / Sample Size

### Example

Population = 100 customers

Desired Sample = 10 customers

Sampling interval:

k = 100 / 10 = 10

We can select every 10th customer.

### Business Example

A quality-control team may inspect every 20th product coming from a production line.

### Advantages

- Simple
- Fast
- Easy to implement

### Limitation

If the data has a hidden pattern, systematic sampling can introduce bias.

```python
population = np.arange(1, 101)

sample_size = 10

interval = len(population) // sample_size

systematic_sample = population[::interval]

print("Sampling Interval:", interval)
print("Systematic Sample:", systematic_sample)
```

## 6. Sampling Bias

### Definition

Sampling Bias occurs when the selected sample does not properly represent the population.

This can lead to incorrect conclusions.

### Example

Suppose a company wants to understand customer satisfaction.

If it surveys only premium customers, the results may not represent all customers.

### Types of Sampling Bias

- Selection Bias
- Non-response Bias
- Voluntary Response Bias
- Undercoverage Bias

### Business Example

Only asking highly satisfied customers for reviews can produce overly positive results.

### AI/ML Example

If a training dataset contains mostly one demographic group, a machine learning model may perform poorly on underrepresented groups.

```python
population = pd.DataFrame({
    "Customer": range(1, 11),
    "Satisfaction": [
        2, 3, 4, 5, 2,
        3, 4, 5, 1, 2
    ]
})

biased_sample = population[
    population["Satisfaction"] >= 4
]

print("Population:")
print(population)

print("\nBiased Sample:")
print(biased_sample)
```

## 7. Sampling Error

### Definition

Sampling Error is the difference between a population statistic and the statistic calculated from a sample.

It occurs because we study only a sample instead of the entire population.

### Example

Population mean:

50

Sample mean:

53

Sampling Error:

53 - 50 = 3

### Important Point

Sampling Error is not necessarily a mistake.

It is a natural consequence of using a sample.

### How to Reduce Sampling Error

- Increase sample size
- Use appropriate sampling techniques
- Select a representative sample

```python
population = np.array([10, 20, 30, 40, 50, 60, 70, 80])

sample = np.array([20, 30, 40, 50])

population_mean = np.mean(population)
sample_mean = np.mean(sample)

sampling_error = sample_mean - population_mean

print("Population Mean:", population_mean)
print("Sample Mean:", sample_mean)
print("Sampling Error:", sampling_error)
```

## 8. Sample Size

### Definition

Sample Size is the number of observations included in a sample.

Choosing an appropriate sample size is important because:

- Very small samples may produce unreliable results.
- Very large samples may require unnecessary time and resources.

### Factors Affecting Sample Size

- Population size
- Variability
- Confidence level
- Margin of error
- Research objective

### Business Example

A company with millions of customers may not need to survey every customer.

A sufficiently large representative sample can provide useful estimates.

### AI/ML Example

A larger and representative dataset generally provides more reliable estimates and can help machine learning models generalize better.

However, more data is not automatically better if it contains poor-quality or biased observations.

```python
population_size = 10000
sample_sizes = [50, 100, 500, 1000]

for size in sample_sizes:
    percentage = (size / population_size) * 100
    print(
        f"Sample Size: {size}, "
        f"Population Percentage: {percentage:.2f}%"
    )
```
