## 1. Correlation

### Definition

Correlation measures the strength and direction of the relationship between two variables.

The correlation coefficient usually ranges from -1 to +1.

### Interpretation

- +1 → Perfect positive correlation
- 0 → No linear correlation
- -1 → Perfect negative correlation

### Example

Suppose:

- Study hours increase
- Exam scores also increase

These variables may have positive correlation.

### Business Example

Advertising spending and sales may have a positive correlation.

### AI/ML Example

Correlation helps us understand relationships between features and can help identify useful or redundant features.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

study_hours = [1, 2, 3, 4, 5]
exam_scores = [50, 55, 65, 75, 85]

correlation = np.corrcoef(study_hours, exam_scores)[0, 1]

print("Correlation:", correlation)
```

### Interpretation

The correlation value is close to +1, which indicates a strong positive relationship.

As study hours increase, exam scores also tend to increase.

## 2. Covariance

### Definition

Covariance measures the direction in which two variables change together.

### Interpretation

- Positive covariance → Variables tend to increase together.
- Negative covariance → One variable tends to increase while the other decreases.
- Covariance near 0 → Little evidence of a linear relationship.

### Important Difference

Correlation is standardized and always lies between -1 and +1.

Covariance is not standardized, so its magnitude depends on the scale of the variables.

### Business Example

Advertising spending and sales may have positive covariance.

### AI/ML Example

Covariance is important in algorithms such as Principal Component Analysis (PCA).

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

covariance_matrix = np.cov(x, y)

print("Covariance Matrix:")
print(covariance_matrix)

print("Covariance between x and y:", covariance_matrix[0, 1])
```

### Interpretation

The covariance is positive, which means the two variables tend to increase together.

## 3. Positive Correlation

### Definition

Positive correlation occurs when two variables tend to move in the same direction.

When X increases, Y tends to increase.

When X decreases, Y tends to decrease.

### Example

Experience and salary may have a positive relationship.

Generally, as experience increases, salary may also increase.

### Business Example

Marketing expenditure and sales may have positive correlation.

### AI/ML Example

Two features may both increase as the underlying business activity increases.

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.scatter(x, y)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Positive Correlation")
plt.show()
```

### Interpretation

The points move upward from left to right.

This indicates a positive relationship between X and Y.

## 4. Negative Correlation

### Definition

Negative correlation occurs when two variables tend to move in opposite directions.

When X increases, Y tends to decrease.

### Example

Price and product demand may have a negative relationship.

As price increases, demand may decrease.

### Business Example

Product price and number of purchases can have negative correlation.

### AI/ML Example

Two features may represent opposite aspects of the same underlying behavior.

```python
x = [1, 2, 3, 4, 5]
y = [10, 8, 6, 4, 2]

plt.scatter(x, y)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Negative Correlation")
plt.show()
```

### Interpretation

The points move downward from left to right.

This indicates a negative relationship between X and Y.

## 5. No Correlation

### Definition

No correlation means there is little or no linear relationship between two variables.

### Example

Shoe size and programming skill would generally not be expected to have a meaningful linear relationship.

### Important Point

No correlation does not always mean that there is absolutely no relationship.

There could be a non-linear relationship that Pearson correlation does not detect well.

```python
x = [1, 2, 3, 4, 5]
y = [7, 2, 9, 3, 8]

correlation = np.corrcoef(x, y)[0, 1]

print("Correlation:", correlation)

plt.scatter(x, y)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("No Clear Linear Correlation")
plt.show()
```

### Interpretation

The points do not follow a clear upward or downward linear pattern.

Therefore, there is no strong linear relationship visible in this example.

## 6. Pearson Correlation

### Definition

Pearson Correlation measures the strength and direction of a linear relationship between two numerical variables.

### Range

- +1 → Perfect positive linear relationship
- 0 → No linear relationship
- -1 → Perfect negative linear relationship

### When to Use

Use Pearson correlation when:

- Variables are numerical.
- You are interested in a linear relationship.
- The data is reasonably suitable for Pearson's assumptions.

### Business Example

Measure the relationship between:

Advertising spending and sales.

### AI/ML Example

Use Pearson correlation to examine relationships between numerical features.

```python
from scipy.stats import pearsonr

x = [10, 20, 30, 40, 50]
y = [15, 25, 35, 45, 55]

correlation, p_value = pearsonr(x, y)

print("Pearson Correlation:", correlation)
print("P-value:", p_value)
```

### Interpretation

A Pearson correlation close to +1 indicates a strong positive linear relationship between the variables.

The p-value can be used to test whether the observed correlation is statistically significant under the chosen significance level.

## 7. Spearman Correlation

### Definition

Spearman Correlation measures the strength and direction of a monotonic relationship using the ranks of the data.

It is useful when:

- Data is ordinal.
- The relationship is monotonic but not necessarily linear.
- Data contains outliers that may affect Pearson correlation.
- Normality assumptions for Pearson are not appropriate.

### Example

Customer satisfaction rankings and product rankings.

### Business Example

Compare customer satisfaction rankings with product popularity rankings.

### AI/ML Example

Spearman correlation can help identify monotonic relationships between features.

```python
from scipy.stats import spearmanr

x = [1, 2, 3, 4, 5]
y = [10, 20, 25, 40, 100]

correlation, p_value = spearmanr(x, y)

print("Spearman Correlation:", correlation)
print("P-value:", p_value)
```

### Interpretation

Spearman correlation evaluates whether the variables generally move in the same or opposite ranked order.

Unlike Pearson correlation, it focuses on ranks rather than directly measuring a linear relationship.

## 8. Correlation Matrix

### Definition

A correlation matrix shows the correlation coefficients between multiple numerical variables.

### Example

Suppose a dataset contains:

- Age
- Income
- Spending Score
- Experience

A correlation matrix allows us to examine the relationship between every pair of variables.

### AI/ML Importance

Correlation matrices are useful during Exploratory Data Analysis (EDA).

They can help identify:

- Strong relationships
- Redundant features
- Potential multicollinearity
- Features that may require further investigation

```python
data = pd.DataFrame({
    "Age": [20, 25, 30, 35, 40],
    "Income": [25000, 30000, 40000, 50000, 60000],
    "Spending_Score": [90, 80, 70, 60, 50],
    "Experience": [1, 3, 5, 8, 10]
})

correlation_matrix = data.corr()

print(correlation_matrix)
```

### Interpretation

Each cell contains the correlation between two variables.

Values close to:

- +1 indicate strong positive relationships.
- -1 indicate strong negative relationships.
- 0 indicate weak or no linear relationships.

## 9. Heatmap

### Definition

A heatmap is a visual representation of numerical values using different levels of intensity.

A correlation heatmap makes it easier to identify strong and weak relationships between features.

### Why Use a Heatmap?

A large correlation matrix can be difficult to read.

A heatmap provides a quick visual overview.

### AI/ML Example

Heatmaps are commonly used during EDA to identify highly correlated features before model training.

```python
import seaborn as sns

plt.figure(figsize=(8, 5))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.show()
```

# Correlation vs Covariance

| Feature | Correlation | Covariance |
|---|---|---|
| Measures | Strength and direction | Direction of relationship |
| Range | -1 to +1 | No fixed range |
| Standardized | Yes | No |
| Easy to compare | Yes | More difficult |
| Used in ML | Feature analysis | PCA and statistical analysis |

### Key Difference

Correlation is standardized covariance.

Therefore, correlation is generally easier to interpret and compare across different variables.

# Business Applications

Correlation analysis can be used to study relationships such as:

- Advertising and sales
- Price and demand
- Experience and salary
- Customer visits and purchases
- Website traffic and revenue
- Marketing spending and conversions

### Important Note

Correlation does not prove causation.

If two variables are correlated, it does not automatically mean that one variable causes the other.
