2.DATA TYPES

A data type tells Python what kind of value is stored in a variable.

Python automatically determines the data type based on the value assigned to the variable. Different types of data are used for different purposes in a program.

Some commonly used Python data types are:

int– used for whole numbers such as customer age or number of transactions.
float– used for decimal numbers such as transaction amount or account balance.
str– used for text such as customer name or product category.
bool– used for True or False values such as whether a transaction is fraudulent.

Data types are important in AI/ML because datasets contain different kinds of information. For example, a customer dataset may contain numerical values, text values, and True/False values. Understanding the data type helps us perform the correct data processing and analysis operations.

EXAMPLE :

Consider a customer transaction dataset.

A customer's age can be stored as an integer, the transaction amount can be stored as a float, the customer's name can be stored as a string, and whether the transaction is fraudulent can be stored as a Boolean value.

Python provides the type() function to identify the data type of a value.

IMPLEMENTATION:

In the following example, different customer transaction values are stored in variables. We will use the type() function to check the data type of each variable.

```python
#Integer: whole number
customer_age = 30

# Float: decimal number
transaction_amount = 25000.50

# String: text value
customer_name = "Prateek"

# Boolean: True or False
is_fraud = False

# Display the values
print(customer_age)
print(transaction_amount)
print(customer_name)
print(is_fraud)

# Check the data types
print(type(customer_age))
print(type(transaction_amount))
print(type(customer_name))
print(type(is_fraud))
```

3.Type Casting

Type casting means converting a value from one data type into another data type.

Python provides built-in functions such as int(), float(), str(), and bool() for type conversion.

Type casting is useful in AI/ML applications because data collected from files, APIs, forms, or databases may not always have the required data type.

For example, a transaction amount may be received as a string from an external source, but it may need to be converted into a numeric value before performing calculations.

EXAMPLE :

Suppose a transaction amount is received as text:

"25000.50"

Before performing mathematical operations, we can convert it into a float.

IMPLEMENTATION :

The following example demonstrates converting values between common Python data types.

```python
transaction_amount = "25000.50"
customer_age = "30"
number_of_transactions = 8

transaction_amount = float(transaction_amount)
customer_age = int(customer_age)
transaction_count = float(number_of_transactions)

print(transaction_amount)
print(customer_age)
print(transaction_count)

print(type(transaction_amount))
print(type(customer_age))
print(type(transaction_count))
```

•⁠  ⁠Assignment operators

5.if / elif / else 
   
if, elif, and else are used to make decisions in Python.

  if checks the first condition.
  elif checks another condition if the previous condition is false.

In AI/ML applications, these statements can be used for simple rule-based decisions, data validation, and preprocessing logic.

Example :

25.Recursive Functions :

A recursive function is a function that calls itself.
    
A recursive function must have a condition that stops the recursion.
    
Recursion can be useful for problems involving hierarchical or nested structures.
    
Example :

Suppose we want to calculate the factorial of a number.

```python
def factorial(number):
    if number == 1:
        return 1

    return number * factorial(number - 1)


result = factorial(5)

print(result)
```

ADVANCED PYTHON :

1.Iterators

An iterator is an object that allows Python to access elements one at a time.
Iterators are useful when working with large amounts of data because values can be processed sequentially.

Example :
Suppose we have several transaction records and want to process them one at a time.

```python
transactions = [10000, 25000, 50000]

transaction_iterator = iter(transactions)

print(next(transaction_iterator))
print(next(transaction_iterator))
print(next(transaction_iterator))
```

2.generators :

A generator produces values one at a time instead of storing all values in memory at once.

Generators are especially useful for processing large datasets.

```python
def transaction_generator():
    transactions = [10000, 25000, 50000]

    for transaction in transactions:
        yield transaction


for transaction in transaction_generator():
    print("Processing:", transaction)
```

3.yield : 
The yield keyword is used inside a generator function.

Instead of returning all values at once, yield produces one value and pauses the function until the next value is requested.

```python
def customer_ids():
    for customer_id in range(101, 106):
        yield customer_id


for customer_id in customer_ids():
    print("Customer ID:", customer_id)
```

4.Decorators :

A decorator is a function that modifies or extends the behavior of another function without changing its original code.

Decorators are commonly used in web applications, logging, authentication, monitoring, and AI/ML pipelines.


```python
def log_function(func):
    def wrapper():
        print("Prediction function started")
        func()
        print("Prediction function completed")

    return wrapper


@log_function
def make_prediction():
    print("Generating prediction")


make_prediction()
```

5.Closures :

A closure occurs when an inner function remembers values from its outer function even after the outer function has finished executing.

Closures can be useful for creating configurable functions.

```python
def create_threshold_checker(threshold):

    def check(score):
        return score >= threshold

    return check


high_confidence = create_threshold_checker(0.8)

print(high_confidence(0.9))
print(high_confidence(0.6))
```

6.global and nonlocal :

global is used when a function needs to modify a variable defined outside the function at the global level.

nonlocal is used when an inner function needs to modify a variable from its enclosing function.

These should be used carefully because excessive use can make large applications difficult to maintain.

```python
prediction_count = 0


def make_prediction():
    global prediction_count

    prediction_count += 1


make_prediction()
make_prediction()

print("Predictions:", prediction_count)
```

Python provides many built-in functions that can be used without installing additional libraries.

These functions help perform common operations such as filtering, sorting, checking values, and working with collections.

Some useful built-in functions include:

- map()
- filter()

1.Map() :

map() applies a function to every item in an iterable.
    
It is useful when we need to transform multiple values.

```python
transaction_amounts = ["10000", "25000", "50000"]

converted_amounts = list(
    map(float, transaction_amounts)
)

print(converted_amounts)
```

2.Filter() :

filter() selects items from an iterable based on a condition.

```python
transaction_amounts = [10000, 25000, 75000, 120000]

high_value = list(
    filter(lambda amount: amount > 50000, transaction_amounts)
)

print(high_value)
```

3.Zip() :

zip() combines elements from multiple iterables based on their positions.

```python
customer_ids = [101, 102, 103]
transaction_amounts = [25000, 45000, 75000]

customer_transactions = zip(
    customer_ids,
    transaction_amounts
)

for customer_id, amount in customer_transactions:
    print(customer_id, amount)
```

4.enumerate() :

enumerate() allows us to access both the index and value while iterating through a collection.
    

```python
transaction_amounts = [25000, 45000, 75000]

for index, amount in enumerate(transaction_amounts):
    print("Transaction", index + 1, ":", amount)
```

5.Sorted() :

sorted() returns a new sorted list from an iterable.


```python
transaction_amounts = [75000, 25000, 120000, 45000]

sorted_amounts = sorted(transaction_amounts)

print(sorted_amounts)
```

6.Any() :

all() returns True only if every value in an iterable is true.
    
It can be useful for validating whether all records satisfy a condition.

```python
transaction_amounts = [10000, 25000, 75000, 45000]

all_positive = all(
    amount > 0
    for amount in transaction_amounts
)

print(all_positive)
```

7.Isinstance() :

isinstance() checks whether a value belongs to a particular data type.
    

```python
transaction_amount = 25000.50

if isinstance(transaction_amount, float):
    print("Transaction amount is a float")
else:
    print("Transaction amount is not a float")
```

8.id() :

id() returns the unique identity of an object during its lifetime.
    

```python
transaction_amounts = [10000, 25000, 50000]

another_reference = transaction_amounts

print(id(transaction_amounts))
print(id(another_reference))
```

9.all() :

all() returns True only if every value in an iterable is true.
    

```python
transaction_amounts = [10000, 25000, 75000, 45000]

all_positive = all(
    amount > 0
    for amount in transaction_amounts
)

print(all_positive)
```
