1.1 Understand the Concept

A variable is a name used to store a value in a Python program.

Variables help us store data and use that data later in the program.

Python does not require us to declare the data type separately. The type is automatically determined from the value assigned to the variable.

```python
customer_age = 30
transaction_amount = 25000
previous_transactions = 8

print(customer_age)
print(transaction_amount)
print(previous_transactions)
```

2.DATA TYPES

A data type tells Python what kind of value is stored in a variable.

Python automatically determines the data type based on the value assigned to the variable. Different types of data are used for different purposes in a program.

Some commonly used Python data types are:

int– used for whole numbers such as customer age or number of transactions.
float– used for decimal numbers such as transaction amount or account balance.
str– used for text such as customer name or product category.
bool– used for True or False values such as whether a transaction is fraudulent.

Data types are important in AI/ML because datasets contain different kinds of information. For example, a customer dataset may contain numerical values, text values, and True/False values. Understanding the data type helps us perform the correct data processing and analysis operations.

EXAMPLE :

Consider a customer transaction dataset.

A customer's age can be stored as an integer, the transaction amount can be stored as a float, the customer's name can be stored as a string, and whether the transaction is fraudulent can be stored as a Boolean value.

Python provides the type() function to identify the data type of a value.

IMPLEMENTATION:

In the following example, different customer transaction values are stored in variables. We will use the type() function to check the data type of each variable.

```python
#Integer: whole number
customer_age = 30

# Float: decimal number
transaction_amount = 25000.50

# String: text value
customer_name = "Prateek"

# Boolean: True or False
is_fraud = False

# Display the values
print(customer_age)
print(transaction_amount)
print(customer_name)
print(is_fraud)

# Check the data types
print(type(customer_age))
print(type(transaction_amount))
print(type(customer_name))
print(type(is_fraud))
```

3.Type Casting

Type casting means converting a value from one data type into another data type.

Python provides built-in functions such as int(), float(), str(), and bool() for type conversion.

Type casting is useful in AI/ML applications because data collected from files, APIs, forms, or databases may not always have the required data type.

For example, a transaction amount may be received as a string from an external source, but it may need to be converted into a numeric value before performing calculations.

EXAMPLE :

Suppose a transaction amount is received as text:

"25000.50"

Before performing mathematical operations, we can convert it into a float.

IMPLEMENTATION :

The following example demonstrates converting values between common Python data types.

```python
transaction_amount = "25000.50"
customer_age = "30"
number_of_transactions = 8

transaction_amount = float(transaction_amount)
customer_age = int(customer_age)
transaction_count = float(number_of_transactions)

print(transaction_amount)
print(customer_age)
print(transaction_count)

print(type(transaction_amount))
print(type(customer_age))
print(type(transaction_count))
```

4.Operators 

Operators are symbols or keywords used to perform operations on values and variables.

Python provides different types of operators, including:

•⁠  ⁠Arithmetic operators
•⁠  ⁠Comparison operators
•⁠  ⁠Logical operators
•⁠  ⁠Assignment operators

Operators are important in AI/ML because calculations and comparisons are performed frequently during data processing, feature engineering, validation, and decision-making.

EXAMPLE :

For a customer transaction, arithmetic operators can calculate totals, comparison operators can check conditions, and logical operators can combine multiple conditions.

IMPLEMENTATION :

The following example demonstrates common operators using customer transaction data.

```python
ransaction_amount = 25000
previous_transactions = 8
customer_age = 30

# Arithmetic operators
total_amount = transaction_amount * previous_transactions
average_amount = transaction_amount / previous_transactions

print(total_amount)
print(average_amount)

# Comparison operators
print(customer_age >= 18)
print(transaction_amount > 10000)

# Logical operators
high_value_transaction = transaction_amount > 10000
adult_customer = customer_age >= 18

print(high_value_transaction and adult_customer)
```

4.Control Flow

Control flow determines the order in which Python statements are executed.

In real-world AI/ML applications, programs often need to make decisions, repeat operations, or skip certain operations depending on the data.

EXAMPLE :

ML system may check whether a customer's transaction amount is high before performing additional fraud analysis.

Suppose a transaction-processing system needs to process transactions based on their amount.

IMPLEMENTATION :

The following example demonstrates using conditions and loops to control the execution of a Python program.

```python
transaction_amount = 75000

if transaction_amount > 50000:
    print("Additional fraud verification required")
else:
    print("Normal transaction processing")
```

5.if / elif / else 
   
if, elif, and else are used to make decisions in Python.

  if checks the first condition.
  elif checks another condition if the previous condition is false.
  else executes when none of the conditions are true.
  
In AI/ML applications, these statements can be used for simple rule-based decisions, data validation, and preprocessing logic.

Example :

Suppose a banking system categorizes a transaction based on its amount.

Implementation :

The following example demonstrates how different transaction amounts can be assigned different risk categories.

```python
transaction_amount = 75000

if transaction_amount >= 100000:
    risk_level = "High"
elif transaction_amount >= 50000:
    risk_level = "Medium"
else:
    risk_level = "Low"

print("Transaction Risk:", risk_level)
```

6.for Loop

A for loop is used to repeat a block of code for every item in a sequence or collection.

In AI/ML, loops can be used to process multiple records, features, predictions, or data items.

Example :

Suppose we have several transaction amounts that need to be processed.

Implementation :

The following example demonstrates processing each transaction amount individually.

```python
transaction_amounts = [25000, 45000, 75000, 120000]

for amount in transaction_amounts:
    print("Processing transaction:", amount)
```

7.while Loop

A while loop repeatedly executes code as long as a specified condition remains true.

It is useful when we do not know exactly how many times an operation needs to run.

Example :

Suppose a system needs to retry processing until the required number of attempts is reached.

```python
attempt = 1

while attempt <= 3:
    print("Processing attempt:", attempt)
    attempt += 1
```

8.break
   
The break statement immediately stops a loop.

In AI/ML applications, it can be useful when a required value is found or when a condition requires processing to stop.

Example :

Suppose we are searching transaction records for a suspicious transaction.

```python
ransaction_amounts = [15000, 22000, 35000, 125000, 18000]

for amount in transaction_amounts:
    if amount > 100000:
        print("Suspicious transaction found:", amount)
        break
```

9.continue

The continue statement skips the current iteration of a loop and moves to the next iteration.

It is useful when certain records should be ignored while processing the remaining data.

Example :

Suppose we want to skip transactions below ₹10,000.

```python
transaction_amounts = [5000, 15000, 25000, 7000, 50000]

for amount in transaction_amounts:
    if amount < 10000:
        continue

    print("Processing transaction:", amount)
```
