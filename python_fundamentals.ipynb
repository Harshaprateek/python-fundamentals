BASICS CONCEPTS
1- Variables
1.1 Understand the Concept

A variable is a name used to store a value in a Python program.

Variables help us store data and use that data later in the program.

Python does not require us to declare the data type separately. The type is automatically determined from the value assigned to the variable.

```python
customer_age = 30
transaction_amount = 25000
previous_transactions = 8

print(customer_age)
print(transaction_amount)
print(previous_transactions)
```

2.DATA TYPES

A data type tells Python what kind of value is stored in a variable.

Python automatically determines the data type based on the value assigned to the variable. Different types of data are used for different purposes in a program.

Some commonly used Python data types are:

int– used for whole numbers such as customer age or number of transactions.
float– used for decimal numbers such as transaction amount or account balance.
str– used for text such as customer name or product category.
bool– used for True or False values such as whether a transaction is fraudulent.

Data types are important in AI/ML because datasets contain different kinds of information. For example, a customer dataset may contain numerical values, text values, and True/False values. Understanding the data type helps us perform the correct data processing and analysis operations.

EXAMPLE :

Consider a customer transaction dataset.

A customer's age can be stored as an integer, the transaction amount can be stored as a float, the customer's name can be stored as a string, and whether the transaction is fraudulent can be stored as a Boolean value.

Python provides the type() function to identify the data type of a value.
